```python
# ============================================================
# TEN FABRICS DATASET (TFD)
# COMPLETE ANALYSIS - CORRECTED TXT PARSER
# ============================================================

import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from PIL import Image


# ============================================================
# 1. DATASET PATH
# ============================================================

DATASET_PATH = Path(
    "/kaggle/input/datasets/saharshakir/ten-fabrics-dataset-tfd/TFD Textile Dataset"
)

print("=" * 70)
print("TEN FABRICS DATASET (TFD) ANALYSIS")
print("=" * 70)

print("\nDataset path:")
print(DATASET_PATH)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

print("Dataset path exists: YES")


# ============================================================
# 2. FIND CLASSES
# ============================================================

classes = sorted([
    x.name
    for x in DATASET_PATH.iterdir()
    if x.is_dir()
])

print("\n" + "=" * 70)
print("1. DATASET STRUCTURE")
print("=" * 70)

print("\nClasses:")
print(classes)

print("\nNumber of classes:", len(classes))


# ============================================================
# 3. COUNT IMAGES AND TXT FILES
# ============================================================

distribution = []

all_images = []
all_txt_files = []

for cls in classes:

    folder = DATASET_PATH / cls

    png_files = sorted(folder.glob("*.png"))
    txt_files = sorted(folder.glob("*.txt"))

    all_images.extend(png_files)
    all_txt_files.extend(txt_files)

    distribution.append({
        "Class": cls,
        "Images": len(png_files),
        "TXT Files": len(txt_files),
        "Image %": round(
            len(png_files) / max(len(all_images), 1) * 100,
            2
        )
    })


distribution_df = pd.DataFrame(distribution)

# Correct percentage after total is known
total_images = len(all_images)

distribution_df["Image %"] = (
    distribution_df["Images"] /
    total_images * 100
).round(2)

print("\nImages and TXT files per class:")
display(distribution_df)

print("\nTotal PNG images:", len(all_images))
print("Total TXT files:", len(all_txt_files))


# ============================================================
# 4. IMAGE / TXT PAIRING
# ============================================================

print("\n" + "=" * 70)
print("2. IMAGE / TXT PAIRING")
print("=" * 70)

missing_txt = []
missing_image = []

for img_path in all_images:

    txt_path = img_path.with_suffix(".txt")

    if not txt_path.exists():
        missing_txt.append(img_path)


for txt_path in all_txt_files:

    img_path = txt_path.with_suffix(".png")

    if not img_path.exists():
        missing_image.append(txt_path)


print("Images without TXT:", len(missing_txt))
print("TXT files without image:", len(missing_image))


# ============================================================
# 5. IMAGE PROPERTIES
# ============================================================

print("\n" + "=" * 70)
print("3. IMAGE PROPERTIES")
print("=" * 70)

image_info = []

for img_path in all_images:

    try:

        with Image.open(img_path) as img:

            image_info.append({
                "Class": img_path.parent.name,
                "Filename": img_path.name,
                "Width": img.width,
                "Height": img.height,
                "Mode": img.mode,
                "Channels": len(img.getbands())
            })

    except Exception as e:

        print("Error:", img_path, e)


image_df = pd.DataFrame(image_info)

print("\nSuccessfully read images:", len(image_df))

print("\nImage dimensions:")

display(
    image_df
    .groupby(["Width", "Height"])
    .size()
    .reset_index(name="Images")
)

print("\nImage modes:")

display(
    image_df["Mode"]
    .value_counts()
    .rename_axis("Mode")
    .reset_index(name="Images")
)

print("\nChannels:")

display(
    image_df["Channels"]
    .value_counts()
    .sort_index()
    .rename_axis("Channels")
    .reset_index(name="Images")
)


# ============================================================
# 6. SAMPLE IMAGE INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("4. PIXEL INFORMATION")
print("=" * 70)

sample_image_path = random.choice(all_images)

sample_image = np.array(
    Image.open(sample_image_path)
)

print("\nSample image:", sample_image_path.name)
print("Shape:", sample_image.shape)
print("Data type:", sample_image.dtype)
print("Minimum pixel:", sample_image.min())
print("Maximum pixel:", sample_image.max())
print("Raw values per image:", sample_image.size)


# ============================================================
# 7. INSPECT ACTUAL TXT FILE FORMAT
# ============================================================

print("\n" + "=" * 70)
print("5. INSPECTING TXT FILE FORMAT")
print("=" * 70)

sample_txt_files = random.sample(
    all_txt_files,
    min(3, len(all_txt_files))
)

for i, txt_path in enumerate(sample_txt_files, 1):

    print("\n" + "-" * 70)
    print(f"SAMPLE TXT {i}")
    print("-" * 70)

    print("File:", txt_path)

    with open(txt_path, "r") as f:
        lines = f.readlines()

    print("Number of lines:", len(lines))

    print("\nFirst 25 lines:")

    for line_no, line in enumerate(lines[:25], 1):
        print(f"{line_no:02d}: {repr(line.rstrip())}")


# ============================================================
# 8. ROBUST TXT PARSER
# ============================================================

def parse_annotation(txt_path):

    """
    Reads the TFD TXT annotation.

    The files contain metadata/header information followed
    by the actual annotation matrix.

    We search for consecutive rows containing 16 numeric
    values rather than assuming the entire TXT file is a
    pure numerical matrix.
    """

    with open(txt_path, "r") as f:
        lines = f.readlines()

    numeric_rows = []

    for line in lines:

        line = line.strip()

        if not line:
            continue

        # Try whitespace-separated values
        parts = line.split()

        try:

            values = [float(x) for x in parts]

        except ValueError:

            continue

        # We are interested in 16-column rows
        if len(values) == 16:

            numeric_rows.append(values)

    # Need 16 rows × 16 columns
    if len(numeric_rows) >= 16:

        # Take the LAST 16 rows.
        # This avoids accidentally treating metadata
        # as part of the mask.
        matrix = np.array(
            numeric_rows[-16:],
            dtype=float
        )

        if matrix.shape == (16, 16):

            return matrix

    return None


# ============================================================
# 9. PARSE ALL TXT FILES
# ============================================================

print("\n" + "=" * 70)
print("6. PARSING TXT ANNOTATIONS")
print("=" * 70)

annotation_info = []

failed_annotations = []

parsed_masks = {}

for txt_path in all_txt_files:

    try:

        mask = parse_annotation(txt_path)

        if mask is None:

            failed_annotations.append(txt_path)
            continue

        parsed_masks[str(txt_path)] = mask

        unique_values = np.unique(mask)

        defective_blocks = np.sum(mask != 0)

        annotation_info.append({

            "Class": txt_path.parent.name,

            "Filename": txt_path.name,

            "Shape": str(mask.shape),

            "Rows": mask.shape[0],

            "Columns": mask.shape[1],

            "Min": mask.min(),

            "Max": mask.max(),

            "Unique Values":
                str(unique_values.tolist()),

            "Defective Blocks":
                int(defective_blocks),

            "Total Blocks":
                int(mask.size),

            "Defective %":
                round(
                    defective_blocks /
                    mask.size * 100,
                    2
                )
        })

    except Exception as e:

        failed_annotations.append(txt_path)


annotation_df = pd.DataFrame(annotation_info)

print(
    "\nSuccessfully parsed TXT files:",
    len(annotation_df)
)

print(
    "Failed TXT files:",
    len(failed_annotations)
)


# ============================================================
# 10. ANNOTATION STRUCTURE
# ============================================================

if len(annotation_df) > 0:

    print("\nAnnotation shapes:")

    display(
        annotation_df[
            "Shape"
        ]
        .value_counts()
        .rename_axis("Shape")
        .reset_index(name="TXT Files")
    )

    print("\nAnnotation value examples:")

    display(
        annotation_df[
            [
                "Min",
                "Max",
                "Unique Values"
            ]
        ].head(10)
    )

else:

    print(
        "\nWARNING:"
        "\nNo annotations could be parsed."
        "\nInspect the TXT output above."
    )


# ============================================================
# 11. CLEAN VS DEFECTIVE
# ============================================================

if len(annotation_df) > 0:

    print("\n" + "=" * 70)
    print("7. CLEAN VS DEFECTIVE")
    print("=" * 70)

    annotation_df["Condition"] = (
        annotation_df["Defective Blocks"]
        .apply(
            lambda x:
            "Defective"
            if x > 0
            else "Clean"
        )
    )

    condition_counts = (
        annotation_df["Condition"]
        .value_counts()
        .rename_axis("Condition")
        .reset_index(name="Images")
    )

    condition_counts["Percentage"] = (
        condition_counts["Images"] /
        condition_counts["Images"].sum() *
        100
    ).round(2)

    display(condition_counts)

else:

    condition_counts = pd.DataFrame()


# ============================================================
# 12. CLEAN / DEFECTIVE BY CLASS
# ============================================================

if len(annotation_df) > 0:

    print("\n" + "=" * 70)
    print("8. CLEAN VS DEFECTIVE BY CLASS")
    print("=" * 70)

    class_condition = pd.crosstab(
        annotation_df["Class"],
        annotation_df["Condition"]
    )

    if "Clean" not in class_condition:
        class_condition["Clean"] = 0

    if "Defective" not in class_condition:
        class_condition["Defective"] = 0

    class_condition = class_condition[
        ["Clean", "Defective"]
    ]

    class_condition["Total"] = (
        class_condition["Clean"] +
        class_condition["Defective"]
    )

    class_condition["Defective %"] = (
        class_condition["Defective"] /
        class_condition["Total"] *
        100
    ).round(2)

    display(class_condition)


# ============================================================
# 13. DEFECT AREA STATISTICS
# ============================================================

if len(annotation_df) > 0:

    print("\n" + "=" * 70)
    print("9. DEFECT AREA STATISTICS")
    print("=" * 70)

    display(
        annotation_df[
            [
                "Defective Blocks",
                "Total Blocks",
                "Defective %"
            ]
        ].describe()
    )


# ============================================================
# 14. CLASS DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(12, 6))

plt.bar(
    distribution_df["Class"],
    distribution_df["Images"]
)

plt.xlabel("Fabric Class")
plt.ylabel("Number of Images")
plt.title("Image Distribution Across Fabric Classes")

plt.tight_layout()
plt.show()


# ============================================================
# 15. CLEAN / DEFECTIVE GRAPH
# ============================================================

if len(condition_counts) > 0:

    plt.figure(figsize=(8, 5))

    plt.bar(
        condition_counts["Condition"],
        condition_counts["Images"]
    )

    plt.xlabel("Condition")
    plt.ylabel("Number of Images")
    plt.title("Clean vs Defective Images")

    plt.tight_layout()
    plt.show()


# ============================================================
# 16. CLEAN / DEFECTIVE PER CLASS GRAPH
# ============================================================

if len(annotation_df) > 0:

    plt.figure(figsize=(12, 6))

    plt.bar(
        class_condition.index,
        class_condition["Clean"],
        label="Clean"
    )

    plt.bar(
        class_condition.index,
        class_condition["Defective"],
        bottom=class_condition["Clean"],
        label="Defective"
    )

    plt.xlabel("Fabric Class")
    plt.ylabel("Number of Images")

    plt.title(
        "Clean vs Defective Images by Fabric Class"
    )

    plt.legend()

    plt.tight_layout()
    plt.show()


# ============================================================
# 17. SHOW ONE IMAGE FROM EACH CLASS
# ============================================================

print("\n" + "=" * 70)
print("10. VISUAL EXAMPLES FROM ALL CLASSES")
print("=" * 70)

fig, axes = plt.subplots(
    2,
    5,
    figsize=(18, 8)
)

for ax, cls in zip(
    axes.flatten(),
    classes
):

    class_images = list(
        (DATASET_PATH / cls).glob("*.png")
    )

    if class_images:

        img_path = random.choice(
            class_images
        )

        img = Image.open(img_path)

        ax.imshow(img)

        ax.set_title(
            f"Class {cls}\n{img_path.name}"
        )

    ax.axis("off")

plt.tight_layout()
plt.show()


# ============================================================
# 18. CLEAN / DEFECTIVE EXAMPLES
# ============================================================

if len(annotation_df) > 0:

    print("\n" + "=" * 70)
    print("11. CLEAN VS DEFECTIVE EXAMPLES")
    print("=" * 70)

    examples = {}

    for cls in classes:

        folder = DATASET_PATH / cls

        clean_example = None
        defective_example = None

        for txt_path in folder.glob("*.txt"):

            mask = parsed_masks.get(
                str(txt_path)
            )

            if mask is None:
                continue

            image_path = (
                txt_path.with_suffix(".png")
            )

            if not image_path.exists():
                continue

            if np.any(mask != 0):

                if defective_example is None:

                    defective_example = (
                        image_path,
                        txt_path,
                        mask
                    )

            else:

                if clean_example is None:

                    clean_example = (
                        image_path,
                        txt_path,
                        mask
                    )

            if (
                clean_example is not None
                and
                defective_example is not None
            ):
                break

        examples[cls] = {
            "Clean": clean_example,
            "Defective": defective_example
        }


    fig, axes = plt.subplots(
        len(classes),
        2,
        figsize=(10, 4 * len(classes))
    )

    for row, cls in enumerate(classes):

        for col, condition in enumerate(
            ["Clean", "Defective"]
        ):

            example = examples[cls][condition]

            if example is not None:

                image_path, txt_path, mask = (
                    example
                )

                image = Image.open(
                    image_path
                )

                axes[row, col].imshow(
                    image
                )

                axes[row, col].set_title(
                    f"Class {cls} - {condition}\n"
                    f"{image_path.name}"
                )

            else:

                axes[row, col].set_title(
                    f"Class {cls} - {condition}\n"
                    "No example"
                )

            axes[row, col].axis("off")

    plt.tight_layout()
    plt.show()


# ============================================================
# 19. DEFECT MASK OVERLAY
# ============================================================

if len(annotation_df) > 0:

    print("\n" + "=" * 70)
    print("12. DEFECT MASK VISUALIZATION")
    print("=" * 70)

    defective_examples = []

    for cls in classes:

        folder = DATASET_PATH / cls

        for txt_path in folder.glob("*.txt"):

            mask = parsed_masks.get(
                str(txt_path)
            )

            if mask is None:
                continue

            if np.any(mask != 0):

                image_path = (
                    txt_path.with_suffix(".png")
                )

                if image_path.exists():

                    defective_examples.append(
                        (
                            cls,
                            image_path,
                            txt_path,
                            mask
                        )
                    )

    if defective_examples:

        cls, image_path, txt_path, mask = (
            random.choice(defective_examples)
        )

        image = np.array(
            Image.open(image_path)
        )

        print("\nClass:", cls)
        print("Image:", image_path.name)
        print("TXT:", txt_path.name)

        print(
            "Image shape:",
            image.shape
        )

        print(
            "Mask shape:",
            mask.shape
        )

        print(
            "Defective blocks:",
            np.sum(mask != 0)
        )

        print(
            "Total blocks:",
            mask.size
        )

        print(
            "Defective percentage:",
            round(
                np.sum(mask != 0) /
                mask.size * 100,
                2
            ),
            "%"
        )

        fig, axes = plt.subplots(
            1,
            3,
            figsize=(16, 5)
        )

        axes[0].imshow(image)

        axes[0].set_title(
            "Original Fabric Image"
        )

        axes[0].axis("off")

        axes[1].imshow(
            mask,
            interpolation="nearest"
        )

        axes[1].set_title(
            "16 × 16 Defect Mask"
        )

        axes[1].axis("off")

        axes[2].imshow(image)

        axes[2].imshow(
            mask,
            interpolation="nearest",
            alpha=0.45
        )

        axes[2].set_title(
            "Image + Defect Mask"
        )

        axes[2].axis("off")

        plt.tight_layout()
        plt.show()


# ============================================================
# 20. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("FINAL DATASET SUMMARY")
print("=" * 70)

print(f"""
Fabric classes:
    {len(classes)}

Classes:
    {classes}

Total images:
    {len(all_images)}

Total TXT annotations:
    {len(all_txt_files)}

Image dimensions:
    256 × 256

Image channels:
    3 (RGB)

Pixel data type:
    uint8

Raw pixel values per image:
    196,608

Successfully parsed annotations:
    {len(annotation_df)}

Failed annotations:
    {len(failed_annotations)}
""")

if len(annotation_df) > 0:

    print(
        "Annotation shapes:"
    )

    print(
        annotation_df["Shape"]
        .value_counts()
        .to_string()
    )

    print("\nClean / defective:")

    print(
        condition_counts.to_string(
            index=False
        )
    )

print("\n" + "=" * 70)
print("ANALYSIS COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# TFD FABRIC CLASSIFICATION - LIGHTWEIGHT MODEL COMPARISON
#   & DEPLOYMENT PIPELINE
#
# Paste this as the SECOND cell in your Kaggle notebook,
# right after the standard Kaggle boilerplate cell.
# ============================================================

# ============================================================
# 1. CONFIGURATION
# ============================================================
CONFIG = {
    # --- reproducibility ---
    "SEED": 42,

    # --- data ---
    "IMG_SIZE": 224,
    "TEST_FRACTION": 0.20,          # overall held-out test
    "VAL_FRACTION_OF_DEV": 0.10,    # validation carved out of the 80% dev set
    "BATCH_SIZE": 32,
    "NUM_WORKERS": 2,

    # --- augmentation (train only) ---
    "AUG_HFLIP_P": 0.5,
    "AUG_VFLIP_P": 0.2,
    "AUG_ROTATION_DEG": 10,
    "AUG_RESIZED_CROP_SCALE": (0.85, 1.0),
    "AUG_COLOR_JITTER": {"brightness": 0.15, "contrast": 0.15, "saturation": 0.1},
    "AUG_AFFINE_TRANSLATE": (0.05, 0.05),

    # --- training ---
    "PHASE1_EPOCHS": 8,     # head-only training
    "PHASE2_EPOCHS": 15,    # fine-tuning with partially unfrozen backbone
    "PHASE1_LR": 1e-3,
    "PHASE2_LR": 1e-4,
    "WEIGHT_DECAY": 1e-4,
    "UNFREEZE_FRACTION": 0.25,   # fraction of (named) params unfrozen in phase 2
    "EARLY_STOPPING_PATIENCE": 6,
    "GRAD_CLIP_NORM": 5.0,

    # --- efficiency measurement ---
    "LATENCY_WARMUP_ITERS": 20,
    "LATENCY_MEASURE_ITERS": 100,
    "THROUGHPUT_BATCH_SIZE": 16,
    "THROUGHPUT_ITERS": 30,

    # --- best model selection weights (must sum to 1.0) ---
    "SELECTION_WEIGHTS": {
        "macro_f1": 0.40,
        "accuracy": 0.20,
        "latency": 0.20,   # lower is better
        "size": 0.10,      # lower is better
        "params": 0.10,    # lower is better
    },

    # --- paths ---
    "WORKING_DIR": "/kaggle/working",
    "MODELS_DIR": "/kaggle/working/models",
    "RESULTS_DIR": "/kaggle/working/results",
    "CHECKPOINTS_DIR": "/kaggle/working/checkpoints",
}

CLASS_FOLDERS = [f"{i:03d}" for i in range(1, 11)]  # 001 ... 010

# ============================================================
# 2. IMPORTS
# ============================================================
import os
import sys
import json
import time
import random
import shutil
import hashlib
import subprocess
import warnings
from collections import defaultdict, OrderedDict

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
from PIL import Image, ImageFile

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision
from torchvision import models, transforms

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_recall_fscore_support,
    confusion_matrix, classification_report
)
from sklearn.utils.class_weight import compute_class_weight

warnings.filterwarnings("ignore")
ImageFile.LOAD_TRUNCATED_IMAGES = True
matplotlib.rcParams["figure.max_open_warning"] = 50

for d in [CONFIG["WORKING_DIR"], CONFIG["MODELS_DIR"], CONFIG["RESULTS_DIR"], CONFIG["CHECKPOINTS_DIR"]]:
    os.makedirs(d, exist_ok=True)

print(f"PyTorch version : {torch.__version__}")
print(f"Torchvision version : {torchvision.__version__}")
print(f"CUDA available : {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU name : {torch.cuda.get_device_name(0)}")
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device : {DEVICE}")

# ============================================================
# 3. REPRODUCIBILITY
# ============================================================
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    # deterministic-ish, without crippling GPU throughput
    torch.backends.cudnn.benchmark = True

set_seed(CONFIG["SEED"])
print(f"Random seed set to: {CONFIG['SEED']}")

# ============================================================
# 4. DATASET DISCOVERY
# ============================================================
def find_dataset_root(search_root="/kaggle/input"):
    """
    Recursively search for a directory whose immediate subdirectories
    include (most of) the class folders 001..010.
    """
    best_match, best_count = None, 0
    for root, dirs, _ in os.walk(search_root):
        dirset = set(dirs)
        count = sum(1 for c in CLASS_FOLDERS if c in dirset)
        if count > best_count:
            best_count, best_match = count, root
        if count == len(CLASS_FOLDERS):
            return root  # perfect match, stop early
    if best_match is None or best_count == 0:
        raise FileNotFoundError(
            "Could not locate the TFD dataset structure (folders 001-010) "
            f"under {search_root}. Make sure the dataset is attached."
        )
    print(f"WARNING: only found {best_count}/10 class folders under {best_match}. Proceeding anyway.")
    return best_match

DATASET_ROOT = find_dataset_root()
print(f"Dataset root discovered at: {DATASET_ROOT}")

# ============================================================
# 5. DATASET VALIDATION
# ============================================================
def validate_and_collect(dataset_root):
    """
    Walk each class folder, validate every PNG, check for a matching
    .txt annotation (informational only - NOT used as a label), and
    build a clean list of valid (path, class_label) records.
    Also computes an md5 hash per image for duplicate detection.
    """
    records = []
    stats = {
        "total_seen": 0,
        "corrupted": 0,
        "missing_txt": 0,
        "unexpected_dims": 0,
        "non_rgb_converted": 0,
        "empty_classes": [],
        "missing_classes": [],
    }

    for cls in CLASS_FOLDERS:
        cls_dir = os.path.join(dataset_root, cls)
        if not os.path.isdir(cls_dir):
            stats["missing_classes"].append(cls)
            continue

        png_files = [f for f in os.listdir(cls_dir) if f.lower().endswith(".png")]
        if len(png_files) == 0:
            stats["empty_classes"].append(cls)
            continue

        for fname in png_files:
            stats["total_seen"] += 1
            fpath = os.path.join(cls_dir, fname)

            # check corresponding txt annotation (informational only)
            txt_path = os.path.splitext(fpath)[0] + ".txt"
            if not os.path.isfile(txt_path):
                stats["missing_txt"] += 1

            # validate image integrity
            try:
                with Image.open(fpath) as im:
                    im.verify()
                with Image.open(fpath) as im:
                    if im.size != (256, 256):
                        stats["unexpected_dims"] += 1
                    if im.mode != "RGB":
                        stats["non_rgb_converted"] += 1
                    img_bytes = im.convert("RGB").tobytes()
                    md5 = hashlib.md5(img_bytes).hexdigest()
            except Exception:
                stats["corrupted"] += 1
                continue

            records.append({"filepath": fpath, "class_label": cls, "md5": md5})

    return records, stats

print("Validating dataset (this may take a moment)...")
records, val_stats = validate_and_collect(DATASET_ROOT)

print("\n--- Dataset Validation Report ---")
print(f"Total images scanned      : {val_stats['total_seen']}")
print(f"Corrupted / unreadable    : {val_stats['corrupted']} (skipped)")
print(f"Missing .txt annotation   : {val_stats['missing_txt']} (informational only)")
print(f"Unexpected dimensions     : {val_stats['unexpected_dims']} (kept, will be resized)")
print(f"Non-RGB images            : {val_stats['non_rgb_converted']} (will be converted to RGB)")
print(f"Missing class folders     : {val_stats['missing_classes']}")
print(f"Empty class folders       : {val_stats['empty_classes']}")

if len(records) == 0:
    raise RuntimeError("No valid images found. Aborting.")

# ============================================================
# 6. DATAFRAME CREATION
# ============================================================
df = pd.DataFrame(records)

# duplicate / leakage check: drop exact-duplicate images (same md5),
# keeping only the first occurrence, to avoid the same physical image
# appearing in more than one split later.
before = len(df)
df = df.drop_duplicates(subset="md5", keep="first").reset_index(drop=True)
after = len(df)
print(f"\nExact-duplicate images removed (by content hash): {before - after}")
print(f"Remaining valid, unique images: {after}")

classes_sorted = sorted(df["class_label"].unique())
class_to_idx = {c: i for i, c in enumerate(classes_sorted)}
idx_to_class = {i: c for c, i in class_to_idx.items()}
df["class_idx"] = df["class_label"].map(class_to_idx)

print(f"Classes found (sorted)    : {classes_sorted}")

# ============================================================
# 7. DATA DISTRIBUTION
# ============================================================
class_counts = df["class_label"].value_counts().sort_index()
print("\nClass distribution (full, valid, de-duplicated dataset):")
print(class_counts)

plt.figure(figsize=(8, 5))
class_counts.plot(kind="bar", color="steelblue")
plt.title("Class Distribution (Before Augmentation)")
plt.xlabel("Fabric Class")
plt.ylabel("Number of Images")
plt.tight_layout()
plt.savefig(os.path.join(CONFIG["RESULTS_DIR"], "class_distribution.png"), dpi=150)
plt.close()

# sample images grid, one per class
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for ax, cls in zip(axes.flatten(), classes_sorted):
    sample_path = df[df["class_label"] == cls]["filepath"].iloc[0]
    with Image.open(sample_path) as im:
        ax.imshow(im.convert("RGB"))
    ax.set_title(f"Class {cls}")
    ax.axis("off")
plt.tight_layout()
plt.savefig(os.path.join(CONFIG["RESULTS_DIR"], "sample_images_per_class.png"), dpi=150)
plt.close()
print("Saved: class_distribution.png, sample_images_per_class.png")

# ============================================================
# 8. TRAIN / VALIDATION / TEST SPLIT (stratified, leak-free)
# ============================================================
dev_df, test_df = train_test_split(
    df, test_size=CONFIG["TEST_FRACTION"], stratify=df["class_idx"],
    random_state=CONFIG["SEED"]
)
train_df, val_df = train_test_split(
    dev_df, test_size=CONFIG["VAL_FRACTION_OF_DEV"], stratify=dev_df["class_idx"],
    random_state=CONFIG["SEED"]
)

train_df = train_df.reset_index(drop=True)
val_df = val_df.reset_index(drop=True)
test_df = test_df.reset_index(drop=True)

# sanity check: no md5 hash appears in more than one split
train_hashes, val_hashes, test_hashes = set(train_df.md5), set(val_df.md5), set(test_df.md5)
overlap = (train_hashes & val_hashes) | (train_hashes & test_hashes) | (val_hashes & test_hashes)
assert len(overlap) == 0, f"Data leakage detected across splits! Overlapping hashes: {len(overlap)}"

print(f"\nSplit sizes -> train: {len(train_df)}, val: {len(val_df)}, test: {len(test_df)}")
print(f"Split percentages -> train: {len(train_df)/len(df):.1%}, "
      f"val: {len(val_df)/len(df):.1%}, test: {len(test_df)/len(df):.1%}")
print("Leakage check passed: no duplicate image (by content hash) crosses split boundaries.")

# ============================================================
# 9. DATA AUGMENTATION
# ============================================================
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.RandomResizedCrop(
        CONFIG["IMG_SIZE"], scale=CONFIG["AUG_RESIZED_CROP_SCALE"]
    ),
    transforms.RandomHorizontalFlip(p=CONFIG["AUG_HFLIP_P"]),
    transforms.RandomVerticalFlip(p=CONFIG["AUG_VFLIP_P"]),
    transforms.RandomRotation(CONFIG["AUG_ROTATION_DEG"]),
    transforms.RandomAffine(degrees=0, translate=CONFIG["AUG_AFFINE_TRANSLATE"]),
    transforms.ColorJitter(**CONFIG["AUG_COLOR_JITTER"]),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

# deterministic preprocessing only - NO augmentation for val/test
eval_transform = transforms.Compose([
    transforms.Resize((CONFIG["IMG_SIZE"], CONFIG["IMG_SIZE"])),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

# ============================================================
# 10. DATALOADERS
# ============================================================
class FabricDataset(Dataset):
    def __init__(self, dataframe, transform):
        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        with Image.open(row["filepath"]) as im:
            im = im.convert("RGB")
        im = self.transform(im)
        return im, int(row["class_idx"]), row["filepath"]

train_ds = FabricDataset(train_df, train_transform)
val_ds = FabricDataset(val_df, eval_transform)
test_ds = FabricDataset(test_df, eval_transform)

PIN_MEMORY = DEVICE.type == "cuda"
PERSISTENT = CONFIG["NUM_WORKERS"] > 0

train_loader = DataLoader(train_ds, batch_size=CONFIG["BATCH_SIZE"], shuffle=True,
                           num_workers=CONFIG["NUM_WORKERS"], pin_memory=PIN_MEMORY,
                           persistent_workers=PERSISTENT, drop_last=True)
val_loader = DataLoader(val_ds, batch_size=CONFIG["BATCH_SIZE"], shuffle=False,
                         num_workers=CONFIG["NUM_WORKERS"], pin_memory=PIN_MEMORY,
                         persistent_workers=PERSISTENT)
test_loader = DataLoader(test_ds, batch_size=CONFIG["BATCH_SIZE"], shuffle=False,
                          num_workers=CONFIG["NUM_WORKERS"], pin_memory=PIN_MEMORY,
                          persistent_workers=PERSISTENT)

# --- class imbalance: class weights computed from the TRAINING split ONLY ---
# Rationale: the dataset is imbalanced (e.g. class 007 has ~5x class 005's count).
# Naively oversampling the minority classes before splitting risks leaking
# duplicated samples across train/val/test. Instead we (a) split first,
# (b) augment only the training set, and (c) use a class-weighted loss so
# the optimizer penalizes minority-class errors more heavily, without
# fabricating new data or touching val/test.
train_labels_arr = train_df["class_idx"].values
class_weights_arr = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(len(classes_sorted)),
    y=train_labels_arr
)
class_weights_tensor = torch.tensor(class_weights_arr, dtype=torch.float32).to(DEVICE)
print(f"\nClass weights (balanced, from training split only): {np.round(class_weights_arr, 3)}")

# ============================================================
# 11. MODEL DEFINITIONS
# ============================================================
MODEL_NAMES = ["mobilenet_v3_small", "efficientnet_b0", "shufflenet_v2_x1_0"]
CLASSIFIER_ATTR = {
    "mobilenet_v3_small": "classifier",
    "efficientnet_b0": "classifier",
    "shufflenet_v2_x1_0": "fc",
}

def _load_pretrained(ctor, weights_enum_name):
    """Try the modern `weights=` API, fall back to legacy `pretrained=True`."""
    try:
        weights_enum = getattr(models, weights_enum_name)
        return ctor(weights=weights_enum.DEFAULT)
    except Exception:
        try:
            return ctor(pretrained=True)
        except Exception:
            return ctor(weights=None)

def build_model(name, num_classes):
    if name == "mobilenet_v3_small":
        m = _load_pretrained(models.mobilenet_v3_small, "MobileNet_V3_Small_Weights")
        in_f = m.classifier[-1].in_features
        m.classifier[-1] = nn.Linear(in_f, num_classes)
    elif name == "efficientnet_b0":
        m = _load_pretrained(models.efficientnet_b0, "EfficientNet_B0_Weights")
        in_f = m.classifier[1].in_features
        m.classifier[1] = nn.Linear(in_f, num_classes)
    elif name == "shufflenet_v2_x1_0":
        m = _load_pretrained(models.shufflenet_v2_x1_0, "ShuffleNet_V2_X1_0_Weights")
        in_f = m.fc.in_features
        m.fc = nn.Linear(in_f, num_classes)
    else:
        raise ValueError(f"Unknown model name: {name}")
    return m

def try_build_model_with_fallback(name, num_classes):
    """Gracefully fall back to another lightweight architecture if the
    requested one is unavailable in the installed torchvision version."""
    fallback_pool = ["mnasnet1_0", "squeezenet1_1", "resnet18"]
    try:
        return build_model(name, num_classes), name
    except Exception as e:
        print(f"WARNING: could not build {name} ({e}). Trying a fallback architecture.")
        for fb in fallback_pool:
            try:
                ctor = getattr(models, fb)
                m = ctor(weights="DEFAULT") if "weights" in ctor.__init__.__code__.co_varnames else ctor(pretrained=True)
                if fb == "resnet18":
                    m.fc = nn.Linear(m.fc.in_features, num_classes)
                    CLASSIFIER_ATTR[fb] = "fc"
                elif fb == "mnasnet1_0":
                    m.classifier[-1] = nn.Linear(m.classifier[-1].in_features, num_classes)
                    CLASSIFIER_ATTR[fb] = "classifier"
                elif fb == "squeezenet1_1":
                    m.classifier[1] = nn.Conv2d(512, num_classes, kernel_size=1)
                    CLASSIFIER_ATTR[fb] = "classifier"
                print(f"Using fallback architecture: {fb}")
                return m, fb
            except Exception:
                continue
        raise RuntimeError("No lightweight architecture could be constructed.")

def freeze_backbone(model, classifier_attr):
    for p in model.parameters():
        p.requires_grad = False
    classifier = getattr(model, classifier_attr)
    for p in classifier.parameters():
        p.requires_grad = True

def unfreeze_last_fraction(model, classifier_attr, fraction):
    """
    Architecture-agnostic fine-tuning strategy: keep the classifier head
    fully trainable and additionally unfreeze the last `fraction` of the
    model's named parameters (by definition order), which approximates
    unfreezing the deepest backbone layers closest to the output.
    """
    freeze_backbone(model, classifier_attr)
    named = list(model.named_parameters())
    n_unfreeze = max(1, int(len(named) * fraction))
    for _, p in named[-n_unfreeze:]:
        p.requires_grad = True

# ============================================================
# 12. TRAINING FUNCTIONS
# ============================================================
def run_epoch(model, loader, criterion, optimizer, scaler, device, train=True):
    model.train() if train else model.eval()
    total_loss, all_preds, all_labels = 0.0, [], []

    context = torch.enable_grad() if train else torch.no_grad()
    with context:
        for images, labels, _ in loader:
            images, labels = images.to(device, non_blocking=True), labels.to(device, non_blocking=True)

            if train:
                optimizer.zero_grad(set_to_none=True)

            use_amp = device.type == "cuda"
            with torch.autocast(device_type=device.type, enabled=use_amp):
                outputs = model(images)
                loss = criterion(outputs, labels)

            if train:
                if use_amp:
                    scaler.scale(loss).backward()
                    scaler.unscale_(optimizer)
                    torch.nn.utils.clip_grad_norm_(model.parameters(), CONFIG["GRAD_CLIP_NORM"])
                    scaler.step(optimizer)
                    scaler.update()
                else:
                    loss.backward()
                    torch.nn.utils.clip_grad_norm_(model.parameters(), CONFIG["GRAD_CLIP_NORM"])
                    optimizer.step()

            total_loss += loss.item() * images.size(0)
            preds = outputs.argmax(dim=1).detach().cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.detach().cpu().numpy())

    avg_loss = total_loss / len(loader.dataset)
    acc = accuracy_score(all_labels, all_preds)
    _, _, f1_macro, _ = precision_recall_fscore_support(all_labels, all_preds, average="macro", zero_division=0)
    return avg_loss, acc, f1_macro

def train_one_model(model_key, num_classes, class_weights_tensor, train_loader, val_loader, device):
    print(f"\n{'='*60}\nTRAINING MODEL: {model_key}\n{'='*60}")
    model, actual_name = try_build_model_with_fallback(model_key, num_classes)
    model = model.to(device)
    classifier_attr = CLASSIFIER_ATTR.get(actual_name, "classifier")

    criterion = nn.CrossEntropyLoss(weight=class_weights_tensor)
    history = defaultdict(list)
    best_val_f1 = -1.0
    epochs_no_improve = 0
    ckpt_path = os.path.join(CONFIG["CHECKPOINTS_DIR"], f"{actual_name}_best.pth")

    def _train_phase(phase_name, epochs, lr, unfreeze_fn):
        nonlocal best_val_f1, epochs_no_improve
        unfreeze_fn()
        trainable_params = [p for p in model.parameters() if p.requires_grad]
        optimizer = optim.AdamW(trainable_params, lr=lr, weight_decay=CONFIG["WEIGHT_DECAY"])
        scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=2)
        scaler = torch.cuda.amp.GradScaler(enabled=(device.type == "cuda"))

        for epoch in range(epochs):
            t0 = time.time()
            tr_loss, tr_acc, tr_f1 = run_epoch(model, train_loader, criterion, optimizer, scaler, device, train=True)
            val_loss, val_acc, val_f1 = run_epoch(model, val_loader, criterion, optimizer, scaler, device, train=False)
            scheduler.step(val_loss)
            epoch_time = time.time() - t0
            current_lr = optimizer.param_groups[0]["lr"]

            history["train_loss"].append(tr_loss)
            history["val_loss"].append(val_loss)
            history["train_acc"].append(tr_acc)
            history["val_acc"].append(val_acc)

            print(f"[{phase_name}] Epoch {epoch+1}/{epochs} | "
                  f"train_loss={tr_loss:.4f} train_acc={tr_acc:.4f} | "
                  f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} val_f1={val_f1:.4f} | "
                  f"lr={current_lr:.2e} | time={epoch_time:.1f}s")

            if val_f1 > best_val_f1:
                best_val_f1 = val_f1
                epochs_no_improve = 0
                torch.save(model.state_dict(), ckpt_path)
            else:
                epochs_no_improve += 1

            if epochs_no_improve >= CONFIG["EARLY_STOPPING_PATIENCE"]:
                print(f"Early stopping triggered in {phase_name} (no improvement for "
                      f"{CONFIG['EARLY_STOPPING_PATIENCE']} epochs).")
                return True  # stopped early
        return False

    stopped = _train_phase(
        "Phase1-HeadOnly", CONFIG["PHASE1_EPOCHS"], CONFIG["PHASE1_LR"],
        lambda: freeze_backbone(model, classifier_attr)
    )
    if not stopped:
        _train_phase(
            "Phase2-FineTune", CONFIG["PHASE2_EPOCHS"], CONFIG["PHASE2_LR"],
            lambda: unfreeze_last_fraction(model, classifier_attr, CONFIG["UNFREEZE_FRACTION"])
        )

    # reload best checkpoint before returning
    model.load_state_dict(torch.load(ckpt_path, map_location=device))
    model.eval()

    # plot curves
    epochs_range = range(1, len(history["train_loss"]) + 1)
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].plot(epochs_range, history["train_loss"], label="train_loss")
    axes[0].plot(epochs_range, history["val_loss"], label="val_loss")
    axes[0].set_title(f"{actual_name} - Loss")
    axes[0].set_xlabel("Epoch"); axes[0].legend()
    axes[1].plot(epochs_range, history["train_acc"], label="train_acc")
    axes[1].plot(epochs_range, history["val_acc"], label="val_acc")
    axes[1].set_title(f"{actual_name} - Accuracy")
    axes[1].set_xlabel("Epoch"); axes[1].legend()
    plt.tight_layout()
    plt.savefig(os.path.join(CONFIG["RESULTS_DIR"], f"training_curves_{actual_name}.png"), dpi=150)
    plt.close()

    return model, actual_name, ckpt_path, best_val_f1

# ============================================================
# 13-15. TRAIN MODELS 1, 2, 3
# ============================================================
trained_models = OrderedDict()
for model_key in MODEL_NAMES:
    model, actual_name, ckpt_path, best_val_f1 = train_one_model(
        model_key, len(classes_sorted), class_weights_tensor, train_loader, val_loader, DEVICE
    )
    trained_models[actual_name] = {"model": model, "ckpt_path": ckpt_path, "best_val_f1": best_val_f1}

# ============================================================
# 16. TEST EVALUATION
# ============================================================
def evaluate_on_test(model, loader, device):
    model.eval()
    all_preds, all_labels, all_paths = [], [], []
    with torch.inference_mode():
        for images, labels, paths in loader:
            images = images.to(device)
            outputs = model(images)
            preds = outputs.argmax(dim=1).cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())
            all_paths.extend(paths)

    acc = accuracy_score(all_labels, all_preds)
    prec_macro, rec_macro, f1_macro, _ = precision_recall_fscore_support(
        all_labels, all_preds, average="macro", zero_division=0)
    prec_weighted, rec_weighted, f1_weighted, _ = precision_recall_fscore_support(
        all_labels, all_preds, average="weighted", zero_division=0)

    report = classification_report(
        all_labels, all_preds, target_names=classes_sorted, zero_division=0, output_dict=True
    )
    cm = confusion_matrix(all_labels, all_preds)

    return {
        "accuracy": acc,
        "precision_macro": prec_macro, "recall_macro": rec_macro, "f1_macro": f1_macro,
        "precision_weighted": prec_weighted, "recall_weighted": rec_weighted, "f1_weighted": f1_weighted,
        "classification_report": report,
        "confusion_matrix": cm,
        "preds": all_preds, "labels": all_labels, "paths": all_paths,
    }

test_results = {}
for name, info in trained_models.items():
    print(f"\nEvaluating {name} on held-out test set...")
    res = evaluate_on_test(info["model"], test_loader, DEVICE)
    test_results[name] = res
    print(f"  accuracy={res['accuracy']:.4f}  macro_f1={res['f1_macro']:.4f}  "
          f"weighted_f1={res['f1_weighted']:.4f}")

# ============================================================
# 17. CONFUSION MATRICES
# ============================================================
def plot_confusion_matrix(cm, class_names, title, save_path):
    plt.figure(figsize=(7, 6))
    plt.imshow(cm, interpolation="nearest", cmap="Blues")
    plt.title(title)
    plt.colorbar()
    tick_marks = np.arange(len(class_names))
    plt.xticks(tick_marks, class_names, rotation=45)
    plt.yticks(tick_marks, class_names)
    thresh = cm.max() / 2.0
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, format(cm[i, j], "d"), ha="center", va="center",
                      color="white" if cm[i, j] > thresh else "black", fontsize=8)
    plt.ylabel("True label")
    plt.xlabel("Predicted label")
    plt.tight_layout()
    plt.savefig(save_path, dpi=150)
    plt.close()

for name, res in test_results.items():
    plot_confusion_matrix(
        res["confusion_matrix"], classes_sorted, f"Confusion Matrix - {name}",
        os.path.join(CONFIG["RESULTS_DIR"], f"confusion_matrix_{name}.png")
    )
    print(f"\nClassification report for {name}:")
    cr_df = pd.DataFrame(res["classification_report"]).transpose()
    print(cr_df.round(4))
    cr_df.to_csv(os.path.join(CONFIG["RESULTS_DIR"], f"classification_report_{name}.csv"))

# ============================================================
# 18. LATENCY AND THROUGHPUT (+ params, size, FLOPs)
# ============================================================
def count_params(model):
    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    return total, trainable

def get_model_size_mb(model):
    tmp_path = os.path.join(CONFIG["WORKING_DIR"], "_tmp_size_check.pth")
    torch.save(model.state_dict(), tmp_path)
    size_mb = os.path.getsize(tmp_path) / (1024 * 1024)
    os.remove(tmp_path)
    return size_mb

def measure_latency(model, device, img_size, warmup, iters, batch_size=1):
    model.eval()
    dummy = torch.randn(batch_size, 3, img_size, img_size, device=device)
    times = []
    with torch.inference_mode():
        for _ in range(warmup):
            _ = model(dummy)
        if device.type == "cuda":
            torch.cuda.synchronize()
        for _ in range(iters):
            start = time.perf_counter()
            _ = model(dummy)
            if device.type == "cuda":
                torch.cuda.synchronize()
            end = time.perf_counter()
            times.append((end - start) * 1000.0)  # ms
    times = np.array(times)
    return {
        "mean_ms": float(np.mean(times)),
        "median_ms": float(np.median(times)),
        "std_ms": float(np.std(times)),
        "p95_ms": float(np.percentile(times, 95)),
    }

def measure_throughput(model, device, img_size, batch_size, iters):
    model.eval()
    dummy = torch.randn(batch_size, 3, img_size, img_size, device=device)
    with torch.inference_mode():
        for _ in range(5):
            _ = model(dummy)
        if device.type == "cuda":
            torch.cuda.synchronize()
        start = time.perf_counter()
        for _ in range(iters):
            _ = model(dummy)
        if device.type == "cuda":
            torch.cuda.synchronize()
        total_time = time.perf_counter() - start
    total_images = batch_size * iters
    return total_images / total_time  # images/sec

def try_get_flops(model, img_size):
    try:
        from thop import profile
    except ImportError:
        try:
            subprocess.run([sys.executable, "-m", "pip", "install", "-q", "thop"],
                            check=True, timeout=60)
            from thop import profile
        except Exception:
            return "Not available"
    try:
        dummy = torch.randn(1, 3, img_size, img_size).to(next(model.parameters()).device)
        macs, _ = profile(model, inputs=(dummy,), verbose=False)
        return f"{macs / 1e9:.3f} GMACs"
    except Exception:
        return "Not available"

efficiency_results = {}
for name, info in trained_models.items():
    model = info["model"].to(DEVICE)
    total_p, trainable_p = count_params(model)
    size_mb = get_model_size_mb(model)
    latency = measure_latency(model, DEVICE, CONFIG["IMG_SIZE"],
                               CONFIG["LATENCY_WARMUP_ITERS"], CONFIG["LATENCY_MEASURE_ITERS"], batch_size=1)
    throughput = measure_throughput(model, DEVICE, CONFIG["IMG_SIZE"],
                                     CONFIG["THROUGHPUT_BATCH_SIZE"], CONFIG["THROUGHPUT_ITERS"])
    flops = try_get_flops(model, CONFIG["IMG_SIZE"])

    efficiency_results[name] = {
        "total_params": total_p, "trainable_params": trainable_p,
        "size_mb": size_mb, "flops": flops, "throughput_imgs_per_sec": throughput,
        **latency,
    }
    print(f"\n{name} efficiency:")
    print(f"  Params: {total_p:,} total | Size: {size_mb:.2f} MB | FLOPs: {flops}")
    print(f"  Latency (bs=1): mean={latency['mean_ms']:.2f}ms median={latency['median_ms']:.2f}ms "
          f"std={latency['std_ms']:.2f}ms p95={latency['p95_ms']:.2f}ms")
    print(f"  Throughput (bs={CONFIG['THROUGHPUT_BATCH_SIZE']}): {throughput:.2f} images/sec")

# ============================================================
# 19. MODEL COMPARISON
# ============================================================
comparison_rows = []
for name in trained_models.keys():
    tr = test_results[name]
    ef = efficiency_results[name]
    comparison_rows.append({
        "Model": name,
        "Accuracy": tr["accuracy"],
        "Macro_Precision": tr["precision_macro"],
        "Macro_Recall": tr["recall_macro"],
        "Macro_F1": tr["f1_macro"],
        "Weighted_F1": tr["f1_weighted"],
        "Total_Params": ef["total_params"],
        "Model_Size_MB": ef["size_mb"],
        "Mean_Latency_ms": ef["mean_ms"],
        "Median_Latency_ms": ef["median_ms"],
        "P95_Latency_ms": ef["p95_ms"],
        "Throughput_imgs_per_sec": ef["throughput_imgs_per_sec"],
        "FLOPs": ef["flops"],
    })

comparison_df = pd.DataFrame(comparison_rows)
comparison_df.to_csv(os.path.join(CONFIG["WORKING_DIR"], "model_comparison.csv"), index=False)
print("\n=== MODEL COMPARISON TABLE ===")
print(comparison_df.round(4).to_string(index=False))

# comparison charts
def bar_compare(metric_col, ylabel, title, filename, lower_is_better=False):
    plt.figure(figsize=(7, 5))
    colors = ["seagreen" if not lower_is_better else "indianred"] * len(comparison_df)
    plt.bar(comparison_df["Model"], comparison_df[metric_col], color="cornflowerblue")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.xticks(rotation=15)
    plt.tight_layout()
    plt.savefig(os.path.join(CONFIG["RESULTS_DIR"], filename), dpi=150)
    plt.close()

bar_compare("Macro_F1", "Macro F1", "Model Comparison - Macro F1", "model_comparison.png")
bar_compare("Accuracy", "Accuracy", "Model Comparison - Accuracy", "accuracy_f1_comparison.png")
bar_compare("Mean_Latency_ms", "Latency (ms)", "Latency Comparison (bs=1)", "latency_comparison.png", lower_is_better=True)
bar_compare("Throughput_imgs_per_sec", "Images/sec", "Throughput Comparison", "throughput_comparison.png")
bar_compare("Model_Size_MB", "Size (MB)", "Model Size / Parameter Comparison", "param_size_comparison.png", lower_is_better=True)

# ============================================================
# 20. BEST MODEL SELECTION
# ============================================================
# Selection rule (documented, deterministic, uses ONLY test-set metrics
# computed after model development was finished on val - no repeated
# test peeking occurred during training):
#   1. Normalize each metric across the 3 models to [0, 1].
#   2. For "higher is better" metrics (macro F1, accuracy) use the value
#      directly; for "lower is better" metrics (latency, size, params)
#      use (1 - normalized value) so higher score = better everywhere.
#   3. Combine with weights from CONFIG["SELECTION_WEIGHTS"], which
#      prioritize Macro F1 first (most meaningful given class imbalance),
#      then accuracy, then deployment efficiency (latency/size/params).
def normalize(series, lower_is_better=False):
    lo, hi = series.min(), series.max()
    if hi == lo:
        norm = pd.Series([1.0] * len(series), index=series.index)
    else:
        norm = (series - lo) / (hi - lo)
    return (1 - norm) if lower_is_better else norm

w = CONFIG["SELECTION_WEIGHTS"]
scores = (
    w["macro_f1"] * normalize(comparison_df["Macro_F1"]) +
    w["accuracy"] * normalize(comparison_df["Accuracy"]) +
    w["latency"] * normalize(comparison_df["Mean_Latency_ms"], lower_is_better=True) +
    w["size"] * normalize(comparison_df["Model_Size_MB"], lower_is_better=True) +
    w["params"] * normalize(comparison_df["Total_Params"], lower_is_better=True)
)
comparison_df["Selection_Score"] = scores
comparison_df = comparison_df.sort_values("Selection_Score", ascending=False).reset_index(drop=True)
comparison_df.to_csv(os.path.join(CONFIG["WORKING_DIR"], "model_comparison.csv"), index=False)

best_row = comparison_df.iloc[0]
best_model_name = best_row["Model"]
best_model = trained_models[best_model_name]["model"]

print(f"\n{'='*60}\nBEST MODEL: {best_model_name}\n{'='*60}")
print(f"REASON: Highest weighted selection score ({best_row['Selection_Score']:.4f}), "
      f"combining Macro F1 ({best_row['Macro_F1']:.4f}, weight={w['macro_f1']}), "
      f"Accuracy ({best_row['Accuracy']:.4f}, weight={w['accuracy']}), "
      f"mean latency ({best_row['Mean_Latency_ms']:.2f}ms, weight={w['latency']}, lower is better), "
      f"model size ({best_row['Model_Size_MB']:.2f}MB, weight={w['size']}, lower is better), "
      f"and parameter count ({int(best_row['Total_Params']):,}, weight={w['params']}, lower is better).")

# ============================================================
# 21. MODEL EXPORT (for website deployment)
# ============================================================
best_state_dict_path = os.path.join(CONFIG["MODELS_DIR"], "best_model.pth")
torch.save(best_model.state_dict(), best_state_dict_path)
print(f"\nSaved: {best_state_dict_path}")

# also copy each individual model's best checkpoint into models/
for name, info in trained_models.items():
    dst = os.path.join(CONFIG["MODELS_DIR"], f"{name}_best.pth")
    shutil.copy(info["ckpt_path"], dst)

# TorchScript export
scripted_path = os.path.join(CONFIG["MODELS_DIR"], "best_model_scripted.pt")
try:
    best_model_cpu = best_model.to("cpu").eval()
    scripted_model = torch.jit.script(best_model_cpu)
    scripted_model.save(scripted_path)
    print(f"Saved TorchScript model: {scripted_path}")
except Exception as e:
    print(f"torch.jit.script failed ({e}), trying torch.jit.trace...")
    try:
        example_input = torch.randn(1, 3, CONFIG["IMG_SIZE"], CONFIG["IMG_SIZE"])
        traced_model = torch.jit.trace(best_model_cpu, example_input)
        traced_model.save(scripted_path)
        print(f"Saved TorchScript (traced) model: {scripted_path}")
    except Exception as e2:
        print(f"TorchScript export failed entirely ({e2}). Skipping - .pth is still available.")

# ONNX export (best-effort, non-fatal if it fails)
onnx_path = os.path.join(CONFIG["MODELS_DIR"], "best_model.onnx")
try:
    example_input = torch.randn(1, 3, CONFIG["IMG_SIZE"], CONFIG["IMG_SIZE"])
    torch.onnx.export(
        best_model_cpu, example_input, onnx_path,
        input_names=["input"], output_names=["output"],
        dynamic_axes={"input": {0: "batch_size"}, "output": {0: "batch_size"}},
        opset_version=13,
    )
    print(f"Saved ONNX model: {onnx_path}")
except Exception as e:
    print(f"ONNX export failed or unsupported in this environment ({e}). "
          f"This is optional - PyTorch/TorchScript model is still available.")

best_model = best_model.to(DEVICE)  # move back for inference demo below

# ============================================================
# 22. CLASS MAPPING + MODEL CONFIG
# ============================================================
class_names_path = os.path.join(CONFIG["WORKING_DIR"], "class_names.json")
with open(class_names_path, "w") as f:
    json.dump({str(i): idx_to_class[i] for i in range(len(classes_sorted))}, f, indent=4)
print(f"Saved: {class_names_path}")

model_config = {
    "architecture": best_model_name,
    "input_size": CONFIG["IMG_SIZE"],
    "normalize_mean": IMAGENET_MEAN,
    "normalize_std": IMAGENET_STD,
    "num_classes": len(classes_sorted),
    "classes_in_order": classes_sorted,
    "preprocessing": "Resize to (input_size, input_size) -> ToTensor -> Normalize(mean, std). "
                      "No augmentation at inference time.",
}
model_config_path = os.path.join(CONFIG["WORKING_DIR"], "model_config.json")
with open(model_config_path, "w") as f:
    json.dump(model_config, f, indent=4)
print(f"Saved: {model_config_path}")

# ============================================================
# 23. WEBSITE-STYLE INFERENCE FUNCTION
# ============================================================
def predict_image(image_path, model=best_model, device=DEVICE, transform=eval_transform,
                   idx_to_class_map=idx_to_class, verbose=True):
    """
    Reusable single-image inference function suitable for a Flask/FastAPI
    backend. Loads an image, applies the exact validation/test
    preprocessing, runs the model, and returns predicted class + probabilities.
    """
    with Image.open(image_path) as im:
        im = im.convert("RGB")
    tensor = transform(im).unsqueeze(0).to(device)

    model.eval()
    with torch.inference_mode():
        logits = model(tensor)
        probs = torch.softmax(logits, dim=1).cpu().numpy()[0]

    pred_idx = int(np.argmax(probs))
    pred_class = idx_to_class_map[pred_idx]
    confidence = float(probs[pred_idx])
    prob_dict = {idx_to_class_map[i]: float(probs[i]) for i in range(len(probs))}

    if verbose:
        print(f"Predicted Fabric: {pred_class}")
        print(f"Confidence: {confidence*100:.2f}%\n")
        print("Class probabilities:")
        for cls_name, p in prob_dict.items():
            print(f"{cls_name}: {p:.4f}")

    return pred_class, confidence, prob_dict

# ============================================================
# SAMPLE INFERENCE ON TEST IMAGES
# ============================================================
best_res = test_results[best_model_name]
sample_idxs = list(range(len(best_res["labels"])))
random.shuffle(sample_idxs)

correct_idxs = [i for i in sample_idxs if best_res["preds"][i] == best_res["labels"][i]]
incorrect_idxs = [i for i in sample_idxs if best_res["preds"][i] != best_res["labels"][i]]
display_idxs = (correct_idxs[:5] + incorrect_idxs[:3])[:8]

if len(display_idxs) > 0:
    n_cols = min(4, len(display_idxs))
    n_rows = int(np.ceil(len(display_idxs) / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(4 * n_cols, 4 * n_rows))
    axes = np.array(axes).reshape(-1)
    for ax, i in zip(axes, display_idxs):
        path = best_res["paths"][i]
        actual = idx_to_class[best_res["labels"][i]]
        pred = idx_to_class[best_res["preds"][i]]
        pred_class, confidence, _ = predict_image(path, verbose=False)
        with Image.open(path) as im:
            ax.imshow(im.convert("RGB"))
        color = "green" if actual == pred else "red"
        ax.set_title(f"Actual: {actual} | Pred: {pred}\nConf: {confidence*100:.1f}%", color=color, fontsize=9)
        ax.axis("off")
    for ax in axes[len(display_idxs):]:
        ax.axis("off")
    plt.tight_layout()
    plt.savefig(os.path.join(CONFIG["RESULTS_DIR"], "sample_inference.png"), dpi=150)
    plt.close()
    print("Saved: sample_inference.png")

# save all test predictions
test_pred_df = pd.DataFrame({
    "filepath": best_res["paths"],
    "actual_class": [idx_to_class[l] for l in best_res["labels"]],
    "predicted_class": [idx_to_class[p] for p in best_res["preds"]],
})
test_pred_df.to_csv(os.path.join(CONFIG["WORKING_DIR"], "test_predictions.csv"), index=False)

# ============================================================
# README
# ============================================================
readme_text = f"""TFD Fabric Classification - Output Directory
=============================================
best_model.pth              - state_dict of the selected best model ({best_model_name})
best_model_scripted.pt      - TorchScript export of the best model (if export succeeded)
best_model.onnx             - ONNX export of the best model (if export succeeded)
class_names.json            - index -> fabric class label mapping
model_config.json           - architecture, input size, normalization stats for inference
model_comparison.csv        - full metric/efficiency comparison across all 3 models
test_predictions.csv        - per-image actual vs predicted class on the held-out test set
results/                    - all plots (distribution, curves, confusion matrices, comparisons)
checkpoints/                - best-validation checkpoint per model (training time)
models/                     - final deployment-ready model files
"""
with open(os.path.join(CONFIG["WORKING_DIR"], "README.txt"), "w") as f:
    f.write(readme_text)

# ============================================================
# FINAL REPORT
# ============================================================
print("\n" + "=" * 60)
print("TFD FABRIC CLASSIFICATION FINAL REPORT")
print("=" * 60)
print(f"\nDataset:")
print(f"Total valid images : {len(df)}")
print(f"Number of classes  : {len(classes_sorted)}")
print(f"Input size         : {CONFIG['IMG_SIZE']}x{CONFIG['IMG_SIZE']}")
print(f"\nTrain images       : {len(train_df)}")
print(f"Validation images  : {len(val_df)}")
print(f"Test images        : {len(test_df)}")

print("\n" + "-" * 40)
print("MODEL RESULTS")
print("-" * 40)
for _, row in comparison_df.iterrows():
    print(f"\nModel: {row['Model']}")
    print(f"  Accuracy       : {row['Accuracy']:.4f}")
    print(f"  Macro F1       : {row['Macro_F1']:.4f}")
    print(f"  Parameters     : {int(row['Total_Params']):,}")
    print(f"  Model Size     : {row['Model_Size_MB']:.2f} MB")
    print(f"  Mean Latency   : {row['Mean_Latency_ms']:.2f} ms")
    print(f"  P95 Latency    : {row['P95_Latency_ms']:.2f} ms")
    print(f"  Throughput     : {row['Throughput_imgs_per_sec']:.2f} images/sec")

print("\n" + "-" * 40)
print("BEST MODEL")
print("-" * 40)
print(f"Model          : {best_model_name}")
print(f"Test Accuracy  : {best_row['Accuracy']:.4f}")
print(f"Test Macro F1  : {best_row['Macro_F1']:.4f}")
print(f"Latency        : {best_row['Mean_Latency_ms']:.2f} ms")
print(f"Throughput     : {best_row['Throughput_imgs_per_sec']:.2f} images/sec")
print(f"Parameters     : {int(best_row['Total_Params']):,}")
print(f"Model Size     : {best_row['Model_Size_MB']:.2f} MB")

print("\n" + "-" * 40)
print("DEPLOYMENT FILES")
print("-" * 40)
for fname in ["best_model.pth", "best_model_scripted.pt", "best_model.onnx",
              "class_names.json", "model_config.json", "model_comparison.csv",
              "test_predictions.csv", "README.txt"]:
    full_path = os.path.join(CONFIG["WORKING_DIR"], fname) if fname != "best_model.pth" and "best_model" not in fname else \
                (os.path.join(CONFIG["MODELS_DIR"], fname) if "best_model" in fname else os.path.join(CONFIG["WORKING_DIR"], fname))
    exists = os.path.isfile(full_path)
    print(f"{fname:30s} {'[OK]' if exists else '[MISSING]'}")

print("\nPipeline complete. All outputs saved under /kaggle/working/")
```

```python
# ============================================================
# TFD FABRIC CLASSIFICATION - LIGHTWEIGHT MODEL COMPARISON
#   & DEPLOYMENT PIPELINE
#
# Paste this as the SECOND cell in your Kaggle notebook,
# right after the standard Kaggle boilerplate cell.
# ============================================================

# ============================================================
# 1. CONFIGURATION
# ============================================================
CONFIG = {
    # --- reproducibility ---
    "SEED": 42,

    # --- data ---
    "IMG_SIZE": 224,
    "TEST_FRACTION": 0.20,          # overall held-out test
    "VAL_FRACTION_OF_DEV": 0.10,    # validation carved out of the 80% dev set
    "BATCH_SIZE": 32,
    "NUM_WORKERS": 2,

    # --- augmentation (train only) ---
    "AUG_HFLIP_P": 0.5,
    "AUG_VFLIP_P": 0.2,
    "AUG_ROTATION_DEG": 10,
    "AUG_RESIZED_CROP_SCALE": (0.85, 1.0),
    "AUG_COLOR_JITTER": {"brightness": 0.15, "contrast": 0.15, "saturation": 0.1},
    "AUG_AFFINE_TRANSLATE": (0.05, 0.05),

    # --- training ---
    "PHASE1_EPOCHS": 8,     # head-only training
    "PHASE2_EPOCHS": 15,    # fine-tuning with partially unfrozen backbone
    "PHASE1_LR": 1e-3,
    "PHASE2_LR": 1e-4,
    "WEIGHT_DECAY": 1e-4,
    "UNFREEZE_FRACTION": 0.25,   # fraction of (named) params unfrozen in phase 2
    "EARLY_STOPPING_PATIENCE": 6,
    "GRAD_CLIP_NORM": 5.0,

    # --- efficiency measurement ---
    "LATENCY_WARMUP_ITERS": 20,
    "LATENCY_MEASURE_ITERS": 100,
    "THROUGHPUT_BATCH_SIZE": 16,
    "THROUGHPUT_ITERS": 30,

    # --- best model selection weights (must sum to 1.0) ---
    "SELECTION_WEIGHTS": {
        "macro_f1": 0.40,
        "accuracy": 0.20,
        "latency": 0.20,   # lower is better
        "size": 0.10,      # lower is better
        "params": 0.10,    # lower is better
    },

    # --- paths ---
    "WORKING_DIR": "/kaggle/working",
    "MODELS_DIR": "/kaggle/working/models",
    "RESULTS_DIR": "/kaggle/working/results",
    "CHECKPOINTS_DIR": "/kaggle/working/checkpoints",
}

CLASS_FOLDERS = [f"{i:03d}" for i in range(1, 11)]  # 001 ... 010

# ============================================================
# 2. IMPORTS
# ============================================================
import os
import sys
import json
import time
import random
import shutil
import hashlib
import subprocess
import warnings
from collections import defaultdict, OrderedDict

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
from PIL import Image, ImageFile

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import torchvision
from torchvision import models, transforms

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_recall_fscore_support,
    confusion_matrix, classification_report
)
from sklearn.utils.class_weight import compute_class_weight

warnings.filterwarnings("ignore")
ImageFile.LOAD_TRUNCATED_IMAGES = True
matplotlib.rcParams["figure.max_open_warning"] = 50

for d in [CONFIG["WORKING_DIR"], CONFIG["MODELS_DIR"], CONFIG["RESULTS_DIR"], CONFIG["CHECKPOINTS_DIR"]]:
    os.makedirs(d, exist_ok=True)

print(f"PyTorch version : {torch.__version__}")
print(f"Torchvision version : {torchvision.__version__}")
print(f"CUDA available : {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU name : {torch.cuda.get_device_name(0)}")
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device : {DEVICE}")

# ============================================================
# 3. REPRODUCIBILITY
# ============================================================
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    # deterministic-ish, without crippling GPU throughput
    torch.backends.cudnn.benchmark = True

set_seed(CONFIG["SEED"])
print(f"Random seed set to: {CONFIG['SEED']}")

# ============================================================
# 4. DATASET DISCOVERY
# ============================================================
def find_dataset_root(search_root="/kaggle/input"):
    """
    Recursively search for a directory whose immediate subdirectories
    include (most of) the class folders 001..010.
    """
    best_match, best_count = None, 0
    for root, dirs, _ in os.walk(search_root):
        dirset = set(dirs)
        count = sum(1 for c in CLASS_FOLDERS if c in dirset)
        if count > best_count:
            best_count, best_match = count, root
        if count == len(CLASS_FOLDERS):
            return root  # perfect match, stop early
    if best_match is None or best_count == 0:
        raise FileNotFoundError(
            "Could not locate the TFD dataset structure (folders 001-010) "
            f"under {search_root}. Make sure the dataset is attached."
        )
    print(f"WARNING: only found {best_count}/10 class folders under {best_match}. Proceeding anyway.")
    return best_match

DATASET_ROOT = find_dataset_root()
print(f"Dataset root discovered at: {DATASET_ROOT}")

# ============================================================
# 5. DATASET VALIDATION
# ============================================================
def validate_and_collect(dataset_root):
    """
    Walk each class folder, validate every PNG, check for a matching
    .txt annotation (informational only - NOT used as a label), and
    build a clean list of valid (path, class_label) records.
    Also computes an md5 hash per image for duplicate detection.
    """
    records = []
    stats = {
        "total_seen": 0,
        "corrupted": 0,
        "missing_txt": 0,
        "unexpected_dims": 0,
        "non_rgb_converted": 0,
        "empty_classes": [],
        "missing_classes": [],
    }

    for cls in CLASS_FOLDERS:
        cls_dir = os.path.join(dataset_root, cls)
        if not os.path.isdir(cls_dir):
            stats["missing_classes"].append(cls)
            continue

        png_files = [f for f in os.listdir(cls_dir) if f.lower().endswith(".png")]
        if len(png_files) == 0:
            stats["empty_classes"].append(cls)
            continue

        for fname in png_files:
            stats["total_seen"] += 1
            fpath = os.path.join(cls_dir, fname)

            # check corresponding txt annotation (informational only)
            txt_path = os.path.splitext(fpath)[0] + ".txt"
            if not os.path.isfile(txt_path):
                stats["missing_txt"] += 1

            # validate image integrity
            try:
                with Image.open(fpath) as im:
                    im.verify()
                with Image.open(fpath) as im:
                    if im.size != (256, 256):
                        stats["unexpected_dims"] += 1
                    if im.mode != "RGB":
                        stats["non_rgb_converted"] += 1
                    img_bytes = im.convert("RGB").tobytes()
                    md5 = hashlib.md5(img_bytes).hexdigest()
            except Exception:
                stats["corrupted"] += 1
                continue

            records.append({"filepath": fpath, "class_label": cls, "md5": md5})

    return records, stats

print("Validating dataset (this may take a moment)...")
records, val_stats = validate_and_collect(DATASET_ROOT)

print("\n--- Dataset Validation Report ---")
print(f"Total images scanned      : {val_stats['total_seen']}")
print(f"Corrupted / unreadable    : {val_stats['corrupted']} (skipped)")
print(f"Missing .txt annotation   : {val_stats['missing_txt']} (informational only)")
print(f"Unexpected dimensions     : {val_stats['unexpected_dims']} (kept, will be resized)")
print(f"Non-RGB images            : {val_stats['non_rgb_converted']} (will be converted to RGB)")
print(f"Missing class folders     : {val_stats['missing_classes']}")
print(f"Empty class folders       : {val_stats['empty_classes']}")

if len(records) == 0:
    raise RuntimeError("No valid images found. Aborting.")

# ============================================================
# 6. DATAFRAME CREATION
# ============================================================
df = pd.DataFrame(records)

# duplicate / leakage check: drop exact-duplicate images (same md5),
# keeping only the first occurrence, to avoid the same physical image
# appearing in more than one split later.
before = len(df)
df = df.drop_duplicates(subset="md5", keep="first").reset_index(drop=True)
after = len(df)
print(f"\nExact-duplicate images removed (by content hash): {before - after}")
print(f"Remaining valid, unique images: {after}")

# --- NEAR-DUPLICATE / SOURCE-LEAKAGE CHECK ---
# An exact md5 match only catches byte-identical files. Fabric/texture
# datasets frequently contain sequential crops or near-identical frames
# taken from the SAME physical swatch/photo (e.g. 001-001.png, 001-002.png).
# Those images can be visually almost indistinguishable, so if one lands in
# train and its near-duplicate lands in test, the model can "recognize the
# swatch" rather than learn the general fabric texture - this is a classic,
# easy-to-miss source of leakage that produces suspiciously perfect scores.
# We use perceptual hashing (pHash) to detect these and group near-duplicate
# images together with a union-find structure, so the SPLIT (below) can keep
# every image in a group on the same side of train/val/test.
try:
    import imagehash
except ImportError:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "imagehash"], check=False)
    try:
        import imagehash
    except ImportError:
        imagehash = None

if imagehash is not None:
    print("\nRunning near-duplicate (perceptual hash) check - this can take a minute...")
    phashes = []
    for fp in df["filepath"]:
        with Image.open(fp) as im:
            phashes.append(imagehash.phash(im.convert("RGB")))

    NEAR_DUP_HAMMING_THRESHOLD = 4  # lower = stricter (only near-identical images grouped)
    n = len(df)
    parent = list(range(n))

    def _find(x):
        while parent[x] != x:
            parent[x] = parent[parent[x]]
            x = parent[x]
        return x

    def _union(a, b):
        ra, rb = _find(a), _find(b)
        if ra != rb:
            parent[ra] = rb

    # only compare images within the same class - a near-duplicate texture
    # crop cannot belong to a different fabric class, and this keeps the
    # comparison tractable (O(n_per_class^2) instead of O(n_total^2))
    by_class = defaultdict(list)
    for i, cls in enumerate(df["class_label"]):
        by_class[cls].append(i)
    for cls, idxs in by_class.items():
        for a in range(len(idxs)):
            for b in range(a + 1, len(idxs)):
                i, j = idxs[a], idxs[b]
                if phashes[i] - phashes[j] <= NEAR_DUP_HAMMING_THRESHOLD:
                    _union(i, j)

    df["group_id"] = [_find(i) for i in range(n)]
    group_sizes = df["group_id"].value_counts()
    n_groups_with_dups = (group_sizes > 1).sum()
    n_flagged = int((group_sizes[group_sizes > 1] - 1).sum())
    print(f"Near-duplicate groups found: {n_groups_with_dups} "
          f"(covering {n_flagged} images that are visually near-identical to another image)")
    print("These groups will be kept entirely within a single split (train, val, OR test) "
          "to prevent leakage - see the group-aware split below.")
else:
    print("\nWARNING: 'imagehash' package unavailable (offline environment?) - "
          "falling back to exact-duplicate-only leakage protection. Install "
          "'imagehash' for a stronger near-duplicate check.")
    df["group_id"] = range(len(df))

classes_sorted = sorted(df["class_label"].unique())
class_to_idx = {c: i for i, c in enumerate(classes_sorted)}
idx_to_class = {i: c for c, i in class_to_idx.items()}
df["class_idx"] = df["class_label"].map(class_to_idx)

print(f"Classes found (sorted)    : {classes_sorted}")

# ============================================================
# 7. DATA DISTRIBUTION
# ============================================================
class_counts = df["class_label"].value_counts().sort_index()
print("\nClass distribution (full, valid, de-duplicated dataset):")
print(class_counts)

plt.figure(figsize=(8, 5))
class_counts.plot(kind="bar", color="steelblue")
plt.title("Class Distribution (Before Augmentation)")
plt.xlabel("Fabric Class")
plt.ylabel("Number of Images")
plt.tight_layout()
plt.savefig(os.path.join(CONFIG["RESULTS_DIR"], "class_distribution.png"), dpi=150)
plt.close()

# sample images grid, one per class
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for ax, cls in zip(axes.flatten(), classes_sorted):
    sample_path = df[df["class_label"] == cls]["filepath"].iloc[0]
    with Image.open(sample_path) as im:
        ax.imshow(im.convert("RGB"))
    ax.set_title(f"Class {cls}")
    ax.axis("off")
plt.tight_layout()
plt.savefig(os.path.join(CONFIG["RESULTS_DIR"], "sample_images_per_class.png"), dpi=150)
plt.close()
print("Saved: class_distribution.png, sample_images_per_class.png")

# ============================================================
# 8. TRAIN / VALIDATION / TEST SPLIT (group-aware, stratified, leak-free)
# ============================================================
# We split at the GROUP level (near-duplicate clusters from section 6), not
# the image level, so that every image in a near-duplicate group ends up on
# the same side of the train/val/test boundary. Plain `train_test_split`
# cannot guarantee this, so we use StratifiedGroupKFold, which keeps groups
# intact while still balancing classes across folds as well as possible.
from sklearn.model_selection import StratifiedGroupKFold

def group_aware_stratified_split(dataframe, test_frac, val_frac_of_dev, seed):
    labels = dataframe["class_idx"].values
    groups = dataframe["group_id"].values

    n_splits_test = max(2, round(1 / test_frac))
    sgkf = StratifiedGroupKFold(n_splits=n_splits_test, shuffle=True, random_state=seed)
    dev_idx, test_idx = next(sgkf.split(dataframe, labels, groups))
    dev_df_ = dataframe.iloc[dev_idx].reset_index(drop=True)
    test_df_ = dataframe.iloc[test_idx].reset_index(drop=True)

    n_splits_val = max(2, round(1 / val_frac_of_dev))
    sgkf2 = StratifiedGroupKFold(n_splits=n_splits_val, shuffle=True, random_state=seed)
    tr_idx, val_idx = next(sgkf2.split(dev_df_, dev_df_["class_idx"].values, dev_df_["group_id"].values))
    train_df_ = dev_df_.iloc[tr_idx].reset_index(drop=True)
    val_df_ = dev_df_.iloc[val_idx].reset_index(drop=True)
    return train_df_, val_df_, test_df_

train_df, val_df, test_df = group_aware_stratified_split(
    df, CONFIG["TEST_FRACTION"], CONFIG["VAL_FRACTION_OF_DEV"], CONFIG["SEED"]
)

# sanity check 1: no md5 hash appears in more than one split
train_hashes, val_hashes, test_hashes = set(train_df.md5), set(val_df.md5), set(test_df.md5)
overlap_md5 = (train_hashes & val_hashes) | (train_hashes & test_hashes) | (val_hashes & test_hashes)
assert len(overlap_md5) == 0, f"Exact-duplicate leakage detected across splits! Overlapping hashes: {len(overlap_md5)}"

# sanity check 2 (stronger): no near-duplicate GROUP appears in more than one split
train_groups, val_groups, test_groups = set(train_df.group_id), set(val_df.group_id), set(test_df.group_id)
overlap_groups = (train_groups & val_groups) | (train_groups & test_groups) | (val_groups & test_groups)
assert len(overlap_groups) == 0, f"Near-duplicate group leakage detected across splits! Overlapping groups: {len(overlap_groups)}"

print(f"\nSplit sizes -> train: {len(train_df)}, val: {len(val_df)}, test: {len(test_df)}")
print(f"Split percentages -> train: {len(train_df)/len(df):.1%}, "
      f"val: {len(val_df)/len(df):.1%}, test: {len(test_df)/len(df):.1%}")
print("Leakage check passed: no duplicate image (by content hash) crosses split boundaries.")

# ============================================================
# 9. DATA AUGMENTATION
# ============================================================
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.RandomResizedCrop(
        CONFIG["IMG_SIZE"], scale=CONFIG["AUG_RESIZED_CROP_SCALE"]
    ),
    transforms.RandomHorizontalFlip(p=CONFIG["AUG_HFLIP_P"]),
    transforms.RandomVerticalFlip(p=CONFIG["AUG_VFLIP_P"]),
    transforms.RandomRotation(CONFIG["AUG_ROTATION_DEG"]),
    transforms.RandomAffine(degrees=0, translate=CONFIG["AUG_AFFINE_TRANSLATE"]),
    transforms.ColorJitter(**CONFIG["AUG_COLOR_JITTER"]),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

# deterministic preprocessing only - NO augmentation for val/test
eval_transform = transforms.Compose([
    transforms.Resize((CONFIG["IMG_SIZE"], CONFIG["IMG_SIZE"])),
    transforms.ToTensor(),
    transforms.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])

# ============================================================
# 10. DATALOADERS
# ============================================================
class FabricDataset(Dataset):
    def __init__(self, dataframe, transform):
        self.df = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        with Image.open(row["filepath"]) as im:
            im = im.convert("RGB")
        im = self.transform(im)
        return im, int(row["class_idx"]), row["filepath"]

train_ds = FabricDataset(train_df, train_transform)
val_ds = FabricDataset(val_df, eval_transform)
test_ds = FabricDataset(test_df, eval_transform)

PIN_MEMORY = DEVICE.type == "cuda"
PERSISTENT = CONFIG["NUM_WORKERS"] > 0

train_loader = DataLoader(train_ds, batch_size=CONFIG["BATCH_SIZE"], shuffle=True,
                           num_workers=CONFIG["NUM_WORKERS"], pin_memory=PIN_MEMORY,
                           persistent_workers=PERSISTENT, drop_last=True)
val_loader = DataLoader(val_ds, batch_size=CONFIG["BATCH_SIZE"], shuffle=False,
                         num_workers=CONFIG["NUM_WORKERS"], pin_memory=PIN_MEMORY,
                         persistent_workers=PERSISTENT)
test_loader = DataLoader(test_ds, batch_size=CONFIG["BATCH_SIZE"], shuffle=False,
                          num_workers=CONFIG["NUM_WORKERS"], pin_memory=PIN_MEMORY,
                          persistent_workers=PERSISTENT)

# --- class imbalance: class weights computed from the TRAINING split ONLY ---
# Rationale: the dataset is imbalanced (e.g. class 007 has ~5x class 005's count).
# Naively oversampling the minority classes before splitting risks leaking
# duplicated samples across train/val/test. Instead we (a) split first,
# (b) augment only the training set, and (c) use a class-weighted loss so
# the optimizer penalizes minority-class errors more heavily, without
# fabricating new data or touching val/test.
train_labels_arr = train_df["class_idx"].values
class_weights_arr = compute_class_weight(
    class_weight="balanced",
    classes=np.arange(len(classes_sorted)),
    y=train_labels_arr
)
class_weights_tensor = torch.tensor(class_weights_arr, dtype=torch.float32).to(DEVICE)
print(f"\nClass weights (balanced, from training split only): {np.round(class_weights_arr, 3)}")

# ============================================================
# 11. MODEL DEFINITIONS
# ============================================================
MODEL_NAMES = ["mobilenet_v3_small", "efficientnet_b0", "shufflenet_v2_x1_0"]
CLASSIFIER_ATTR = {
    "mobilenet_v3_small": "classifier",
    "efficientnet_b0": "classifier",
    "shufflenet_v2_x1_0": "fc",
}

def _load_pretrained(ctor, weights_enum_name):
    """Try the modern `weights=` API, fall back to legacy `pretrained=True`."""
    try:
        weights_enum = getattr(models, weights_enum_name)
        return ctor(weights=weights_enum.DEFAULT)
    except Exception:
        try:
            return ctor(pretrained=True)
        except Exception:
            return ctor(weights=None)

def build_model(name, num_classes):
    if name == "mobilenet_v3_small":
        m = _load_pretrained(models.mobilenet_v3_small, "MobileNet_V3_Small_Weights")
        in_f = m.classifier[-1].in_features
        m.classifier[-1] = nn.Linear(in_f, num_classes)
    elif name == "efficientnet_b0":
        m = _load_pretrained(models.efficientnet_b0, "EfficientNet_B0_Weights")
        in_f = m.classifier[1].in_features
        m.classifier[1] = nn.Linear(in_f, num_classes)
    elif name == "shufflenet_v2_x1_0":
        m = _load_pretrained(models.shufflenet_v2_x1_0, "ShuffleNet_V2_X1_0_Weights")
        in_f = m.fc.in_features
        m.fc = nn.Linear(in_f, num_classes)
    else:
        raise ValueError(f"Unknown model name: {name}")
    return m

def try_build_model_with_fallback(name, num_classes):
    """Gracefully fall back to another lightweight architecture if the
    requested one is unavailable in the installed torchvision version."""
    fallback_pool = ["mnasnet1_0", "squeezenet1_1", "resnet18"]
    try:
        return build_model(name, num_classes), name
    except Exception as e:
        print(f"WARNING: could not build {name} ({e}). Trying a fallback architecture.")
        for fb in fallback_pool:
            try:
                ctor = getattr(models, fb)
                m = ctor(weights="DEFAULT") if "weights" in ctor.__init__.__code__.co_varnames else ctor(pretrained=True)
                if fb == "resnet18":
                    m.fc = nn.Linear(m.fc.in_features, num_classes)
                    CLASSIFIER_ATTR[fb] = "fc"
                elif fb == "mnasnet1_0":
                    m.classifier[-1] = nn.Linear(m.classifier[-1].in_features, num_classes)
                    CLASSIFIER_ATTR[fb] = "classifier"
                elif fb == "squeezenet1_1":
                    m.classifier[1] = nn.Conv2d(512, num_classes, kernel_size=1)
                    CLASSIFIER_ATTR[fb] = "classifier"
                print(f"Using fallback architecture: {fb}")
                return m, fb
            except Exception:
                continue
        raise RuntimeError("No lightweight architecture could be constructed.")

def freeze_backbone(model, classifier_attr):
    for p in model.parameters():
        p.requires_grad = False
    classifier = getattr(model, classifier_attr)
    for p in classifier.parameters():
        p.requires_grad = True

def unfreeze_last_fraction(model, classifier_attr, fraction):
    """
    Architecture-agnostic fine-tuning strategy: keep the classifier head
    fully trainable and additionally unfreeze the last `fraction` of the
    model's named parameters (by definition order), which approximates
    unfreezing the deepest backbone layers closest to the output.
    """
    freeze_backbone(model, classifier_attr)
    named = list(model.named_parameters())
    n_unfreeze = max(1, int(len(named) * fraction))
    for _, p in named[-n_unfreeze:]:
        p.requires_grad = True

# ============================================================
# 12. TRAINING FUNCTIONS
# ============================================================
def run_epoch(model, loader, criterion, optimizer, scaler, device, train=True):
    model.train() if train else model.eval()
    total_loss, all_preds, all_labels = 0.0, [], []

    context = torch.enable_grad() if train else torch.no_grad()
    with context:
        for images, labels, _ in loader:
            images, labels = images.to(device, non_blocking=True), labels.to(device, non_blocking=True)

            if train:
                optimizer.zero_grad(set_to_none=True)

            use_amp = device.type == "cuda"
            with torch.autocast(device_type=device.type, enabled=use_amp):
                outputs = model(images)
                loss = criterion(outputs, labels)

            if train:
                if use_amp:
                    scaler.scale(loss).backward()
                    scaler.unscale_(optimizer)
                    torch.nn.utils.clip_grad_norm_(model.parameters(), CONFIG["GRAD_CLIP_NORM"])
                    scaler.step(optimizer)
                    scaler.update()
                else:
                    loss.backward()
                    torch.nn.utils.clip_grad_norm_(model.parameters(), CONFIG["GRAD_CLIP_NORM"])
                    optimizer.step()

            total_loss += loss.item() * images.size(0)
            preds = outputs.argmax(dim=1).detach().cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.detach().cpu().numpy())

    avg_loss = total_loss / len(loader.dataset)
    acc = accuracy_score(all_labels, all_preds)
    _, _, f1_macro, _ = precision_recall_fscore_support(all_labels, all_preds, average="macro", zero_division=0)
    return avg_loss, acc, f1_macro

def train_one_model(model_key, num_classes, class_weights_tensor, train_loader, val_loader, device):
    print(f"\n{'='*60}\nTRAINING MODEL: {model_key}\n{'='*60}")
    model, actual_name = try_build_model_with_fallback(model_key, num_classes)
    model = model.to(device)
    classifier_attr = CLASSIFIER_ATTR.get(actual_name, "classifier")

    criterion = nn.CrossEntropyLoss(weight=class_weights_tensor)
    history = defaultdict(list)
    best_val_f1 = -1.0
    epochs_no_improve = 0
    ckpt_path = os.path.join(CONFIG["CHECKPOINTS_DIR"], f"{actual_name}_best.pth")

    def _train_phase(phase_name, epochs, lr, unfreeze_fn):
        nonlocal best_val_f1, epochs_no_improve
        unfreeze_fn()
        trainable_params = [p for p in model.parameters() if p.requires_grad]
        optimizer = optim.AdamW(trainable_params, lr=lr, weight_decay=CONFIG["WEIGHT_DECAY"])
        scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode="min", factor=0.5, patience=2)
        scaler = torch.cuda.amp.GradScaler(enabled=(device.type == "cuda"))

        for epoch in range(epochs):
            t0 = time.time()
            tr_loss, tr_acc, tr_f1 = run_epoch(model, train_loader, criterion, optimizer, scaler, device, train=True)
            val_loss, val_acc, val_f1 = run_epoch(model, val_loader, criterion, optimizer, scaler, device, train=False)
            scheduler.step(val_loss)
            epoch_time = time.time() - t0
            current_lr = optimizer.param_groups[0]["lr"]

            history["train_loss"].append(tr_loss)
            history["val_loss"].append(val_loss)
            history["train_acc"].append(tr_acc)
            history["val_acc"].append(val_acc)

            print(f"[{phase_name}] Epoch {epoch+1}/{epochs} | "
                  f"train_loss={tr_loss:.4f} train_acc={tr_acc:.4f} | "
                  f"val_loss={val_loss:.4f} val_acc={val_acc:.4f} val_f1={val_f1:.4f} | "
                  f"lr={current_lr:.2e} | time={epoch_time:.1f}s")

            if val_f1 > best_val_f1:
                best_val_f1 = val_f1
                epochs_no_improve = 0
                torch.save(model.state_dict(), ckpt_path)
            else:
                epochs_no_improve += 1

            if epochs_no_improve >= CONFIG["EARLY_STOPPING_PATIENCE"]:
                print(f"Early stopping triggered in {phase_name} (no improvement for "
                      f"{CONFIG['EARLY_STOPPING_PATIENCE']} epochs).")
                return True  # stopped early
        return False

    stopped = _train_phase(
        "Phase1-HeadOnly", CONFIG["PHASE1_EPOCHS"], CONFIG["PHASE1_LR"],
        lambda: freeze_backbone(model, classifier_attr)
    )
    if not stopped:
        _train_phase(
            "Phase2-FineTune", CONFIG["PHASE2_EPOCHS"], CONFIG["PHASE2_LR"],
            lambda: unfreeze_last_fraction(model, classifier_attr, CONFIG["UNFREEZE_FRACTION"])
        )

    # reload best checkpoint before returning
    model.load_state_dict(torch.load(ckpt_path, map_location=device))
    model.eval()

    # plot curves
    epochs_range = range(1, len(history["train_loss"]) + 1)
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))
    axes[0].plot(epochs_range, history["train_loss"], label="train_loss")
    axes[0].plot(epochs_range, history["val_loss"], label="val_loss")
    axes[0].set_title(f"{actual_name} - Loss")
    axes[0].set_xlabel("Epoch"); axes[0].legend()
    axes[1].plot(epochs_range, history["train_acc"], label="train_acc")
    axes[1].plot(epochs_range, history["val_acc"], label="val_acc")
    axes[1].set_title(f"{actual_name} - Accuracy")
    axes[1].set_xlabel("Epoch"); axes[1].legend()
    plt.tight_layout()
    plt.savefig(os.path.join(CONFIG["RESULTS_DIR"], f"training_curves_{actual_name}.png"), dpi=150)
    plt.close()

    return model, actual_name, ckpt_path, best_val_f1

# ============================================================
# 13-15. TRAIN MODELS 1, 2, 3
# ============================================================
trained_models = OrderedDict()
for model_key in MODEL_NAMES:
    model, actual_name, ckpt_path, best_val_f1 = train_one_model(
        model_key, len(classes_sorted), class_weights_tensor, train_loader, val_loader, DEVICE
    )
    trained_models[actual_name] = {"model": model, "ckpt_path": ckpt_path, "best_val_f1": best_val_f1}

# ============================================================
# 16. VALIDATION EVALUATION (for model selection) + TEST EVALUATION (report only)
# ============================================================
def evaluate_model(model, loader, device):
    model.eval()
    all_preds, all_labels, all_paths = [], [], []
    with torch.inference_mode():
        for images, labels, paths in loader:
            images = images.to(device)
            outputs = model(images)
            preds = outputs.argmax(dim=1).cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())
            all_paths.extend(paths)

    acc = accuracy_score(all_labels, all_preds)
    prec_macro, rec_macro, f1_macro, _ = precision_recall_fscore_support(
        all_labels, all_preds, average="macro", zero_division=0)
    prec_weighted, rec_weighted, f1_weighted, _ = precision_recall_fscore_support(
        all_labels, all_preds, average="weighted", zero_division=0)

    report = classification_report(
        all_labels, all_preds, target_names=classes_sorted, zero_division=0, output_dict=True
    )
    cm = confusion_matrix(all_labels, all_preds)

    return {
        "accuracy": acc,
        "precision_macro": prec_macro, "recall_macro": rec_macro, "f1_macro": f1_macro,
        "precision_weighted": prec_weighted, "recall_weighted": rec_weighted, "f1_weighted": f1_weighted,
        "classification_report": report,
        "confusion_matrix": cm,
        "preds": all_preds, "labels": all_labels, "paths": all_paths,
    }

# --- VALIDATION metrics: used ONLY to pick the winning architecture (section 20) ---
val_results = {}
for name, info in trained_models.items():
    val_results[name] = evaluate_model(info["model"], val_loader, DEVICE)

# --- TEST metrics: computed exactly once per model, AFTER training is fully
# finished, purely for reporting. These numbers play NO role in choosing the
# best model (see section 20) - that decision is made from val_results above,
# so the test set stays a genuinely unseen, unbiased final check. ---
test_results = {}
for name, info in trained_models.items():
    print(f"\nEvaluating {name} on held-out test set (final reporting only)...")
    res = evaluate_model(info["model"], test_loader, DEVICE)
    test_results[name] = res
    print(f"  accuracy={res['accuracy']:.4f}  macro_f1={res['f1_macro']:.4f}  "
          f"weighted_f1={res['f1_weighted']:.4f}")

# ============================================================
# 17. CONFUSION MATRICES
# ============================================================
def plot_confusion_matrix(cm, class_names, title, save_path):
    plt.figure(figsize=(7, 6))
    plt.imshow(cm, interpolation="nearest", cmap="Blues")
    plt.title(title)
    plt.colorbar()
    tick_marks = np.arange(len(class_names))
    plt.xticks(tick_marks, class_names, rotation=45)
    plt.yticks(tick_marks, class_names)
    thresh = cm.max() / 2.0
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            plt.text(j, i, format(cm[i, j], "d"), ha="center", va="center",
                      color="white" if cm[i, j] > thresh else "black", fontsize=8)
    plt.ylabel("True label")
    plt.xlabel("Predicted label")
    plt.tight_layout()
    plt.savefig(save_path, dpi=150)
    plt.close()

for name, res in test_results.items():
    plot_confusion_matrix(
        res["confusion_matrix"], classes_sorted, f"Confusion Matrix - {name}",
        os.path.join(CONFIG["RESULTS_DIR"], f"confusion_matrix_{name}.png")
    )
    print(f"\nClassification report for {name}:")
    cr_df = pd.DataFrame(res["classification_report"]).transpose()
    print(cr_df.round(4))
    cr_df.to_csv(os.path.join(CONFIG["RESULTS_DIR"], f"classification_report_{name}.csv"))

# ============================================================
# 18. LATENCY AND THROUGHPUT (+ params, size, FLOPs)
# ============================================================
def count_params(model):
    total = sum(p.numel() for p in model.parameters())
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    return total, trainable

def get_model_size_mb(model):
    tmp_path = os.path.join(CONFIG["WORKING_DIR"], "_tmp_size_check.pth")
    torch.save(model.state_dict(), tmp_path)
    size_mb = os.path.getsize(tmp_path) / (1024 * 1024)
    os.remove(tmp_path)
    return size_mb

def measure_latency(model, device, img_size, warmup, iters, batch_size=1):
    model.eval()
    dummy = torch.randn(batch_size, 3, img_size, img_size, device=device)
    times = []
    with torch.inference_mode():
        for _ in range(warmup):
            _ = model(dummy)
        if device.type == "cuda":
            torch.cuda.synchronize()
        for _ in range(iters):
            start = time.perf_counter()
            _ = model(dummy)
            if device.type == "cuda":
                torch.cuda.synchronize()
            end = time.perf_counter()
            times.append((end - start) * 1000.0)  # ms
    times = np.array(times)
    return {
        "mean_ms": float(np.mean(times)),
        "median_ms": float(np.median(times)),
        "std_ms": float(np.std(times)),
        "p95_ms": float(np.percentile(times, 95)),
    }

def measure_throughput(model, device, img_size, batch_size, iters):
    model.eval()
    dummy = torch.randn(batch_size, 3, img_size, img_size, device=device)
    with torch.inference_mode():
        for _ in range(5):
            _ = model(dummy)
        if device.type == "cuda":
            torch.cuda.synchronize()
        start = time.perf_counter()
        for _ in range(iters):
            _ = model(dummy)
        if device.type == "cuda":
            torch.cuda.synchronize()
        total_time = time.perf_counter() - start
    total_images = batch_size * iters
    return total_images / total_time  # images/sec

def try_get_flops(model, img_size):
    try:
        from thop import profile
    except ImportError:
        try:
            subprocess.run([sys.executable, "-m", "pip", "install", "-q", "thop"],
                            check=True, timeout=60)
            from thop import profile
        except Exception:
            return "Not available"
    try:
        dummy = torch.randn(1, 3, img_size, img_size).to(next(model.parameters()).device)
        macs, _ = profile(model, inputs=(dummy,), verbose=False)
        return f"{macs / 1e9:.3f} GMACs"
    except Exception:
        return "Not available"

efficiency_results = {}
for name, info in trained_models.items():
    model = info["model"].to(DEVICE)
    total_p, trainable_p = count_params(model)
    size_mb = get_model_size_mb(model)
    latency = measure_latency(model, DEVICE, CONFIG["IMG_SIZE"],
                               CONFIG["LATENCY_WARMUP_ITERS"], CONFIG["LATENCY_MEASURE_ITERS"], batch_size=1)
    throughput = measure_throughput(model, DEVICE, CONFIG["IMG_SIZE"],
                                     CONFIG["THROUGHPUT_BATCH_SIZE"], CONFIG["THROUGHPUT_ITERS"])
    flops = try_get_flops(model, CONFIG["IMG_SIZE"])

    efficiency_results[name] = {
        "total_params": total_p, "trainable_params": trainable_p,
        "size_mb": size_mb, "flops": flops, "throughput_imgs_per_sec": throughput,
        **latency,
    }
    print(f"\n{name} efficiency:")
    print(f"  Params: {total_p:,} total | Size: {size_mb:.2f} MB | FLOPs: {flops}")
    print(f"  Latency (bs=1): mean={latency['mean_ms']:.2f}ms median={latency['median_ms']:.2f}ms "
          f"std={latency['std_ms']:.2f}ms p95={latency['p95_ms']:.2f}ms")
    print(f"  Throughput (bs={CONFIG['THROUGHPUT_BATCH_SIZE']}): {throughput:.2f} images/sec")

# ============================================================
# 19. MODEL COMPARISON
# ============================================================
# The table includes BOTH validation and test metrics for transparency,
# but only the Val_* columns feed into model selection (section 20) -
# Test_* columns are here purely for the final, unbiased report.
comparison_rows = []
for name in trained_models.keys():
    vr = val_results[name]
    tr = test_results[name]
    ef = efficiency_results[name]
    comparison_rows.append({
        "Model": name,
        "Val_Accuracy": vr["accuracy"],
        "Val_Macro_F1": vr["f1_macro"],
        "Test_Accuracy": tr["accuracy"],
        "Test_Macro_Precision": tr["precision_macro"],
        "Test_Macro_Recall": tr["recall_macro"],
        "Test_Macro_F1": tr["f1_macro"],
        "Test_Weighted_F1": tr["f1_weighted"],
        "Total_Params": ef["total_params"],
        "Model_Size_MB": ef["size_mb"],
        "Mean_Latency_ms": ef["mean_ms"],
        "Median_Latency_ms": ef["median_ms"],
        "P95_Latency_ms": ef["p95_ms"],
        "Throughput_imgs_per_sec": ef["throughput_imgs_per_sec"],
        "FLOPs": ef["flops"],
    })

comparison_df = pd.DataFrame(comparison_rows)
comparison_df.to_csv(os.path.join(CONFIG["WORKING_DIR"], "model_comparison.csv"), index=False)
print("\n=== MODEL COMPARISON TABLE (Val used for selection, Test for final reporting) ===")
print(comparison_df.round(4).to_string(index=False))

# comparison charts (test metrics, for the final report visuals)
def bar_compare(metric_col, ylabel, title, filename, lower_is_better=False):
    plt.figure(figsize=(7, 5))
    plt.bar(comparison_df["Model"], comparison_df[metric_col], color="cornflowerblue")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.xticks(rotation=15)
    plt.tight_layout()
    plt.savefig(os.path.join(CONFIG["RESULTS_DIR"], filename), dpi=150)
    plt.close()

bar_compare("Test_Macro_F1", "Macro F1 (test)", "Model Comparison - Macro F1", "model_comparison.png")
bar_compare("Test_Accuracy", "Accuracy (test)", "Model Comparison - Accuracy", "accuracy_f1_comparison.png")
bar_compare("Mean_Latency_ms", "Latency (ms)", "Latency Comparison (bs=1)", "latency_comparison.png", lower_is_better=True)
bar_compare("Throughput_imgs_per_sec", "Images/sec", "Throughput Comparison", "throughput_comparison.png")
bar_compare("Model_Size_MB", "Size (MB)", "Model Size / Parameter Comparison", "param_size_comparison.png", lower_is_better=True)

# ============================================================
# 20. BEST MODEL SELECTION
# ============================================================
# Selection rule (documented, deterministic, uses ONLY VALIDATION metrics
# + hardware efficiency metrics - the test set is NEVER consulted when
# choosing which of the 3 architectures wins, per the leakage-prevention
# rule: "use the validation set for model selection; evaluate the test
# set only after model selection"):
#   1. Normalize each metric across the 3 models to [0, 1].
#   2. For "higher is better" metrics (val macro F1, val accuracy) use the
#      value directly; for "lower is better" metrics (latency, size,
#      params - which are hardware/architecture properties, not
#      data-dependent, so using them poses no leakage risk) use
#      (1 - normalized value) so higher score = better everywhere.
#   3. Combine with weights from CONFIG["SELECTION_WEIGHTS"], which
#      prioritize Macro F1 first (most meaningful given class imbalance),
#      then accuracy, then deployment efficiency (latency/size/params).
def normalize(series, lower_is_better=False):
    lo, hi = series.min(), series.max()
    if hi == lo:
        norm = pd.Series([1.0] * len(series), index=series.index)
    else:
        norm = (series - lo) / (hi - lo)
    return (1 - norm) if lower_is_better else norm

w = CONFIG["SELECTION_WEIGHTS"]
scores = (
    w["macro_f1"] * normalize(comparison_df["Val_Macro_F1"]) +
    w["accuracy"] * normalize(comparison_df["Val_Accuracy"]) +
    w["latency"] * normalize(comparison_df["Mean_Latency_ms"], lower_is_better=True) +
    w["size"] * normalize(comparison_df["Model_Size_MB"], lower_is_better=True) +
    w["params"] * normalize(comparison_df["Total_Params"], lower_is_better=True)
)
comparison_df["Selection_Score"] = scores
comparison_df = comparison_df.sort_values("Selection_Score", ascending=False).reset_index(drop=True)
comparison_df.to_csv(os.path.join(CONFIG["WORKING_DIR"], "model_comparison.csv"), index=False)

best_row = comparison_df.iloc[0]
best_model_name = best_row["Model"]
best_model = trained_models[best_model_name]["model"]

print(f"\n{'='*60}\nBEST MODEL: {best_model_name}\n{'='*60}")
print(f"REASON: Highest weighted selection score ({best_row['Selection_Score']:.4f}), computed "
      f"from VALIDATION metrics only, combining Val Macro F1 ({best_row['Val_Macro_F1']:.4f}, "
      f"weight={w['macro_f1']}), Val Accuracy ({best_row['Val_Accuracy']:.4f}, weight={w['accuracy']}), "
      f"mean latency ({best_row['Mean_Latency_ms']:.2f}ms, weight={w['latency']}, lower is better), "
      f"model size ({best_row['Model_Size_MB']:.2f}MB, weight={w['size']}, lower is better), "
      f"and parameter count ({int(best_row['Total_Params']):,}, weight={w['params']}, lower is better). "
      f"Test-set performance (Test Macro F1={best_row['Test_Macro_F1']:.4f}, "
      f"Test Accuracy={best_row['Test_Accuracy']:.4f}) is reported below for reference only "
      f"and played no role in this decision.")

# ============================================================
# 21. MODEL EXPORT (for website deployment)
# ============================================================
best_state_dict_path = os.path.join(CONFIG["MODELS_DIR"], "best_model.pth")
torch.save(best_model.state_dict(), best_state_dict_path)
print(f"\nSaved: {best_state_dict_path}")

# also copy each individual model's best checkpoint into models/
for name, info in trained_models.items():
    dst = os.path.join(CONFIG["MODELS_DIR"], f"{name}_best.pth")
    shutil.copy(info["ckpt_path"], dst)

# TorchScript export
scripted_path = os.path.join(CONFIG["MODELS_DIR"], "best_model_scripted.pt")
try:
    best_model_cpu = best_model.to("cpu").eval()
    scripted_model = torch.jit.script(best_model_cpu)
    scripted_model.save(scripted_path)
    print(f"Saved TorchScript model: {scripted_path}")
except Exception as e:
    print(f"torch.jit.script failed ({e}), trying torch.jit.trace...")
    try:
        example_input = torch.randn(1, 3, CONFIG["IMG_SIZE"], CONFIG["IMG_SIZE"])
        traced_model = torch.jit.trace(best_model_cpu, example_input)
        traced_model.save(scripted_path)
        print(f"Saved TorchScript (traced) model: {scripted_path}")
    except Exception as e2:
        print(f"TorchScript export failed entirely ({e2}). Skipping - .pth is still available.")

# ONNX export (best-effort, non-fatal if it fails)
onnx_path = os.path.join(CONFIG["MODELS_DIR"], "best_model.onnx")
try:
    example_input = torch.randn(1, 3, CONFIG["IMG_SIZE"], CONFIG["IMG_SIZE"])
    torch.onnx.export(
        best_model_cpu, example_input, onnx_path,
        input_names=["input"], output_names=["output"],
        dynamic_axes={"input": {0: "batch_size"}, "output": {0: "batch_size"}},
        opset_version=13,
    )
    print(f"Saved ONNX model: {onnx_path}")
except Exception as e:
    print(f"ONNX export failed or unsupported in this environment ({e}). "
          f"This is optional - PyTorch/TorchScript model is still available.")

best_model = best_model.to(DEVICE)  # move back for inference demo below

# ============================================================
# 22. CLASS MAPPING + MODEL CONFIG
# ============================================================
class_names_path = os.path.join(CONFIG["WORKING_DIR"], "class_names.json")
with open(class_names_path, "w") as f:
    json.dump({str(i): idx_to_class[i] for i in range(len(classes_sorted))}, f, indent=4)
print(f"Saved: {class_names_path}")

model_config = {
    "architecture": best_model_name,
    "input_size": CONFIG["IMG_SIZE"],
    "normalize_mean": IMAGENET_MEAN,
    "normalize_std": IMAGENET_STD,
    "num_classes": len(classes_sorted),
    "classes_in_order": classes_sorted,
    "preprocessing": "Resize to (input_size, input_size) -> ToTensor -> Normalize(mean, std). "
                      "No augmentation at inference time.",
}
model_config_path = os.path.join(CONFIG["WORKING_DIR"], "model_config.json")
with open(model_config_path, "w") as f:
    json.dump(model_config, f, indent=4)
print(f"Saved: {model_config_path}")

# ============================================================
# 23. WEBSITE-STYLE INFERENCE FUNCTION
# ============================================================
def predict_image(image_path, model=best_model, device=DEVICE, transform=eval_transform,
                   idx_to_class_map=idx_to_class, verbose=True):
    """
    Reusable single-image inference function suitable for a Flask/FastAPI
    backend. Loads an image, applies the exact validation/test
    preprocessing, runs the model, and returns predicted class + probabilities.
    """
    with Image.open(image_path) as im:
        im = im.convert("RGB")
    tensor = transform(im).unsqueeze(0).to(device)

    model.eval()
    with torch.inference_mode():
        logits = model(tensor)
        probs = torch.softmax(logits, dim=1).cpu().numpy()[0]

    pred_idx = int(np.argmax(probs))
    pred_class = idx_to_class_map[pred_idx]
    confidence = float(probs[pred_idx])
    prob_dict = {idx_to_class_map[i]: float(probs[i]) for i in range(len(probs))}

    if verbose:
        print(f"Predicted Fabric: {pred_class}")
        print(f"Confidence: {confidence*100:.2f}%\n")
        print("Class probabilities:")
        for cls_name, p in prob_dict.items():
            print(f"{cls_name}: {p:.4f}")

    return pred_class, confidence, prob_dict

# ============================================================
# SAMPLE INFERENCE ON TEST IMAGES
# ============================================================
best_res = test_results[best_model_name]
sample_idxs = list(range(len(best_res["labels"])))
random.shuffle(sample_idxs)

correct_idxs = [i for i in sample_idxs if best_res["preds"][i] == best_res["labels"][i]]
incorrect_idxs = [i for i in sample_idxs if best_res["preds"][i] != best_res["labels"][i]]
display_idxs = (correct_idxs[:5] + incorrect_idxs[:3])[:8]

if len(display_idxs) > 0:
    n_cols = min(4, len(display_idxs))
    n_rows = int(np.ceil(len(display_idxs) / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(4 * n_cols, 4 * n_rows))
    axes = np.array(axes).reshape(-1)
    for ax, i in zip(axes, display_idxs):
        path = best_res["paths"][i]
        actual = idx_to_class[best_res["labels"][i]]
        pred = idx_to_class[best_res["preds"][i]]
        pred_class, confidence, _ = predict_image(path, verbose=False)
        with Image.open(path) as im:
            ax.imshow(im.convert("RGB"))
        color = "green" if actual == pred else "red"
        ax.set_title(f"Actual: {actual} | Pred: {pred}\nConf: {confidence*100:.1f}%", color=color, fontsize=9)
        ax.axis("off")
    for ax in axes[len(display_idxs):]:
        ax.axis("off")
    plt.tight_layout()
    plt.savefig(os.path.join(CONFIG["RESULTS_DIR"], "sample_inference.png"), dpi=150)
    plt.close()
    print("Saved: sample_inference.png")

# save all test predictions
test_pred_df = pd.DataFrame({
    "filepath": best_res["paths"],
    "actual_class": [idx_to_class[l] for l in best_res["labels"]],
    "predicted_class": [idx_to_class[p] for p in best_res["preds"]],
})
test_pred_df.to_csv(os.path.join(CONFIG["WORKING_DIR"], "test_predictions.csv"), index=False)

# ============================================================
# README
# ============================================================
readme_text = f"""TFD Fabric Classification - Output Directory
=============================================
best_model.pth              - state_dict of the selected best model ({best_model_name})
best_model_scripted.pt      - TorchScript export of the best model (if export succeeded)
best_model.onnx             - ONNX export of the best model (if export succeeded)
class_names.json            - index -> fabric class label mapping
model_config.json           - architecture, input size, normalization stats for inference
model_comparison.csv        - full metric/efficiency comparison across all 3 models
test_predictions.csv        - per-image actual vs predicted class on the held-out test set
results/                    - all plots (distribution, curves, confusion matrices, comparisons)
checkpoints/                - best-validation checkpoint per model (training time)
models/                     - final deployment-ready model files
"""
with open(os.path.join(CONFIG["WORKING_DIR"], "README.txt"), "w") as f:
    f.write(readme_text)

# ============================================================
# FINAL REPORT
# ============================================================
print("\n" + "=" * 60)
print("TFD FABRIC CLASSIFICATION FINAL REPORT")
print("=" * 60)
print(f"\nDataset:")
print(f"Total valid images : {len(df)}")
print(f"Number of classes  : {len(classes_sorted)}")
print(f"Input size         : {CONFIG['IMG_SIZE']}x{CONFIG['IMG_SIZE']}")
print(f"\nTrain images       : {len(train_df)}")
print(f"Validation images  : {len(val_df)}")
print(f"Test images        : {len(test_df)}")

print("\n" + "-" * 40)
print("MODEL RESULTS")
print("-" * 40)
for _, row in comparison_df.iterrows():
    print(f"\nModel: {row['Model']}")
    print(f"  Val Accuracy   : {row['Val_Accuracy']:.4f}  (used for selection)")
    print(f"  Val Macro F1   : {row['Val_Macro_F1']:.4f}  (used for selection)")
    print(f"  Test Accuracy  : {row['Test_Accuracy']:.4f}  (final report only)")
    print(f"  Test Macro F1  : {row['Test_Macro_F1']:.4f}  (final report only)")
    print(f"  Parameters     : {int(row['Total_Params']):,}")
    print(f"  Model Size     : {row['Model_Size_MB']:.2f} MB")
    print(f"  Mean Latency   : {row['Mean_Latency_ms']:.2f} ms")
    print(f"  P95 Latency    : {row['P95_Latency_ms']:.2f} ms")
    print(f"  Throughput     : {row['Throughput_imgs_per_sec']:.2f} images/sec")

print("\n" + "-" * 40)
print("BEST MODEL")
print("-" * 40)
print(f"Model          : {best_model_name}")
print(f"Test Accuracy  : {best_row['Test_Accuracy']:.4f}")
print(f"Test Macro F1  : {best_row['Test_Macro_F1']:.4f}")
print(f"Latency        : {best_row['Mean_Latency_ms']:.2f} ms")
print(f"Throughput     : {best_row['Throughput_imgs_per_sec']:.2f} images/sec")
print(f"Parameters     : {int(best_row['Total_Params']):,}")
print(f"Model Size     : {best_row['Model_Size_MB']:.2f} MB")

print("\n" + "-" * 40)
print("DEPLOYMENT FILES")
print("-" * 40)
for fname in ["best_model.pth", "best_model_scripted.pt", "best_model.onnx",
              "class_names.json", "model_config.json", "model_comparison.csv",
              "test_predictions.csv", "README.txt"]:
    full_path = os.path.join(CONFIG["WORKING_DIR"], fname) if fname != "best_model.pth" and "best_model" not in fname else \
                (os.path.join(CONFIG["MODELS_DIR"], fname) if "best_model" in fname else os.path.join(CONFIG["WORKING_DIR"], fname))
    exists = os.path.isfile(full_path)
    print(f"{fname:30s} {'[OK]' if exists else '[MISSING]'}")

print("\nPipeline complete. All outputs saved under /kaggle/working/")
```
